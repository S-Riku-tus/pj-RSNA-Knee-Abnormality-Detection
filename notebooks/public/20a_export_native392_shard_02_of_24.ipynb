```python
from pathlib import Path
import base64, hashlib, io, json, os, sys, zipfile
if os.getenv('KAGGLE_IS_COMPETITION_RERUN'):
    raise RuntimeError('Pixel cache exports must never be submitted for scoring')
CODE = Path('/kaggle/working/native392-shard-code')
if CODE.exists():
    raise RuntimeError('Use a fresh immutable saved run')
payload = base64.b64decode('', validate=True)
if hashlib.sha256(payload).hexdigest() != '3c46c2e5587a83875aaddae22bbecb17413c474bb73a523a90dbfeacdaee77f4':
    raise ValueError('Embedded code ZIP hash differs')
expected = {'src/rsna_knee/__init__.py': 'a6f06061df2e890790f44c5ba4af472dd0abc6f465ad48044e6cee53ae0f7751', 'src/rsna_knee/contracts.py': 'aa246fd1a49ecbe88f1795f49904ef9675f119c935e99b9ffd5eee8220cbc71b', 'src/rsna_knee/imaging.py': '07a114888cde19ae0e5799f2de48c6751ec9c4b5637ec1f4f4c220f2911db9ff', 'src/rsna_knee/feature_contracts.py': 'be76ecc28b14491f4d1a0a459794c7f94456e845ce0e46c46ced699bfaf54ed6', 'src/rsna_knee/feature_imaging.py': 'de36a21929b482eee897b12e06bb9964cee59beab6a9bc2b2d44c15ace279546', 'src/rsna_knee/triplet_cache.py': '0b4c666407cb72d7ef3249f479af1e13cca1b4422b18f18cebf3eed2b768f26f', 'src/rsna_knee/native_pixel_shards.py': 'aa7830e36e65377b0dbdc5857cbcbd2e0261fc13c7ccaeb825c099a47e5f387a', 'bounded_feature_reader.py': '95a906f7e157cf1fa6d722432e6ed913838c6f579ac352418255688a5cf49ee6', 'decoder_runtime.py': '054e6d7e03c602832b37a7ab0a0253aec08847e363afc90e6ba2418e1318210e', 'native392-plan.json': 'd33bab6decbb978d553ae8e75dff7a0bb10276283fcc0511cf86a8441fe9917f'}
with zipfile.ZipFile(io.BytesIO(payload)) as archive:
    if len(archive.infolist()) != len(expected) or set(archive.namelist()) != set(expected):
        raise ValueError('Code/structure-only member allowlist differs')
    validated = {}
    for name in expected:
        target = (CODE / name).resolve()
        if not target.is_relative_to(CODE.resolve()) or '\\' in name or ':' in name:
            raise ValueError('Unsafe code ZIP path')
        data = archive.read(name)
        if hashlib.sha256(data).hexdigest() != expected[name]:
            raise ValueError('Code member bytes differ')
        validated[name] = data
    CODE.mkdir()
    for name, data in validated.items():
        target = CODE / name
        target.parent.mkdir(parents=True, exist_ok=True)
        with target.open('xb') as stream:
            stream.write(data)
sys.path.insert(0, str(CODE))
sys.path.insert(0, str(CODE / 'src'))
CSV_OPENED = []
def prohibit_teacher_csv(event, args):
    if event == 'open' and args and isinstance(args[0], (str, bytes)):
        path = os.fsdecode(args[0])
        if Path(path).name.lower().endswith('.csv') and '/kaggle/input/' in path.replace('\\', '/'):
            if Path(path).name.lower() != 'train_series.csv':
                raise RuntimeError('Teacher, Report, test and submission CSV reads are forbidden')
            CSV_OPENED.append(path)
sys.addaudithook(prohibit_teacher_csv)
print('Code/structural UID plan verified; no teacher or encoder loaded')
```
