```python
%%writefile /kaggle/working/p010_src/p006_metrics.py
"""Observation helpers embedded only in the p006 diagnostic notebook.

No MRI, labels, weights or Kaggle APIs are accessed by this file locally.
"""
import contextlib
import hashlib
import io
import json
import shutil
import subprocess
import sys
import threading
import time
from pathlib import Path


def write_json(path, obj):
    Path(path).write_text(json.dumps(obj, ensure_ascii=False, indent=2) + "\n", encoding="utf-8")


def sha256(path):
    with Path(path).open("rb") as stream:
        return hashlib.file_digest(stream, "sha256").hexdigest()


class Resources:
    """Sample process-tree RSS and device memory, without changing predictions."""

    def __init__(self):
        import psutil
        self.proc = psutil.Process()
        self.started = None
        self.stop = threading.Event()
        self.peak_rss = 0
        self.gpu_peak_mib = []
        self.samples = 0
        self.thread = threading.Thread(target=self._loop, daemon=True)

    def _sample(self):
        import psutil
        rss = 0
        for process in [self.proc, *self.proc.children(recursive=True)]:
            try:
                rss += process.memory_info().rss
            except (psutil.NoSuchProcess, psutil.AccessDenied):
                pass
        self.peak_rss = max(self.peak_rss, rss)
        # RSS is a sum of per-process resident pages; shared pages may be counted twice.
        try:
            completed = subprocess.run(
                ["nvidia-smi", "--query-gpu=memory.used", "--format=csv,noheader,nounits"],
                capture_output=True, text=True, timeout=5, check=True,
            )
            values = [int(x.strip()) for x in completed.stdout.splitlines() if x.strip()]
            if not self.gpu_peak_mib:
                self.gpu_peak_mib = values
            else:
                self.gpu_peak_mib = [max(a, b) for a, b in zip(self.gpu_peak_mib, values)]
        except (OSError, ValueError, subprocess.SubprocessError):
            pass
        self.samples += 1

    def _loop(self):
        while not self.stop.wait(0.5):
            self._sample()

    def __enter__(self):
        self.started = time.monotonic()
        self._sample()
        self.thread.start()
        return self

    def __exit__(self, *args):
        self.stop.set()
        self.thread.join(timeout=10)
        self._sample()
        self.elapsed = time.monotonic() - self.started

    def result(self):
        return {
            "elapsed_seconds": self.elapsed,
            "sampled_peak_process_tree_rss_bytes": self.peak_rss,
            "rss_note": "Sum of RSS includes shared pages more than once; sampled, not exact peak.",
            "sampled_gpu_peak_used_mib": self.gpu_peak_mib,
            "gpu_note": "nvidia-smi device-wide sampled memory, includes unrelated allocations on that device.",
            "samples": self.samples,
            "disk_free_bytes_after": shutil.disk_usage("/kaggle/working").free,
        }


@contextlib.contextmanager
def reader_observer(dataset_class, k, output_root="/kaggle/working"):
    """Record exact returned masks; return the original tensors unchanged."""
    directory = Path(output_root) / f"p006_cov_k{k}"
    directory.mkdir(exist_ok=False)
    original = dataset_class.__getitem__

    def observed(self, index):
        result = original(self, index)
        x, mask, pos = result
        record = {
            "uid": str(self.studies[index]),
            "k": k,
            "shape": list(x.shape),
            "mask": mask.tolist(),
            "unique_positions_per_slot": [len(set(v)) for v in pos.tolist()],
        }
        write_json(directory / f"{index:06d}.json", record)
        return result

    dataset_class.__getitem__ = observed
    try:
        yield
    finally:
        dataset_class.__getitem__ = original


def check_csv(path, uids, labels):
    import numpy as np
    import pandas as pd
    frame = pd.read_csv(path, dtype={"StudyInstanceUID": str})
    assert list(frame.columns) == ["StudyInstanceUID", *labels], (path, frame.columns)
    assert frame.StudyInstanceUID.tolist() == uids, (path, "UID drift")
    values = frame[labels].to_numpy()
    assert np.isfinite(values).all(), (path, "nonfinite")
    assert ((values >= 0) & (values <= 1)).all(), (path, "range")
    assert (frame[labels].nunique() > 1).all(), (path, "constant")
    return {"sha256": sha256(path), "rows": len(frame), "columns": list(frame.columns)}


def paired_readers(namespace, fusion_source, infer_source, uids, labels):
    """Run the unchanged fusion cell twice, observing k16 and k24 only."""
    import numpy as np
    import pandas as pd
    import torch
    reports = {}
    origin = "    run(data_dir, split, ckpts).to_csv(out_csv)"
    assert infer_source.count(origin) == 1
    signature = "def run(data_dir, split, ckpts, k=16, studies=None, workers=4, bs=4):"
    assert infer_source.count(signature) == 1
    path = Path("/kaggle/working/own_src/infer.py")
    for k in (16, 24):
        shutil.copyfile("/kaggle/working/_sota_0949.csv", "/kaggle/working/submission.csv")
        Path("/kaggle/working/_own.csv").unlink(missing_ok=True)
        changed = infer_source.replace(signature, signature.replace("k=16", f"k={k}"))
        # Observer only: model loading, batch size, preprocessing and model calls are unchanged.
        invocation = (
            "    from p006_metrics import reader_observer\n"
            f"    with reader_observer(DicomStudyDataset, {k}):\n"
            "        run(data_dir, split, ckpts).to_csv(out_csv)"
        )
        changed = changed.replace(origin, invocation)
        path.write_text(changed, encoding="utf-8")
        captured = io.StringIO()
        destination = sys.stdout

        class Tee:
            def write(self, text):
                destination.write(text)
                return captured.write(text)

            def flush(self):
                destination.flush()

        with Resources() as resources, contextlib.redirect_stdout(Tee()):
            exec(compile(fusion_source, f"p006-fusion-k{k}", "exec"), namespace)
        log = captured.getvalue()
        assert log.count("[Apex Fusion SUCCESS] Successfully fused") == 1
        assert "[Apex Fusion EXCEPTION]" not in log and "[Apex Fusion FALLBACK]" not in log
        own = Path("/kaggle/working/_own.csv")
        fused = Path("/kaggle/working/submission.csv")
        own_record = check_csv(own, uids, labels)
        fused_record = check_csv(fused, uids, labels)
        # A fallback must not be mistaken for a successfully tested reader.
        expected = pd.read_csv("/kaggle/working/_sota_0949.csv")[labels].rank(pct=True)
        reader = pd.read_csv(own)[labels].rank(pct=True)
        for label in labels:
            w = namespace["TARGET_WEIGHTS"][label]
            expected[label] = (1 - w) * expected[label] + w * reader[label]
        expected = expected.rank(pct=True)
        got = pd.read_csv(fused)[labels]
        assert np.allclose(got, expected, atol=1e-12, rtol=0), "Fusion failed or differs"
        coverage = [json.loads(p.read_text(encoding="utf-8"))
                    for p in sorted(Path(f"/kaggle/working/p006_cov_k{k}").glob("*.json"))]
        assert [row["uid"] for row in coverage] == uids
        assert all(row["shape"] == [6, k, 3, 384, 384] for row in coverage)
        assert all(any(row["mask"]) for row in coverage), "Study has no valid ConvNeXt slots"
        reports[str(k)] = {
            **resources.result(), "reader_csv": own_record, "fused_csv": fused_record,
            "coverage": coverage, "fusion_matches_fixed_formula": True,
            "fusion_success_marker": True, "fusion_exception_or_fallback": False,
        }
        shutil.copyfile(own, f"/kaggle/working/p006_reader_k{k}.csv")
        shutil.copyfile(fused, f"/kaggle/working/p006_fused_k{k}.csv")
        torch.cuda.empty_cache()
    assert [r["mask"] for r in reports["16"]["coverage"]] == [r["mask"] for r in reports["24"]["coverage"]]
    left = pd.read_csv("/kaggle/working/p006_reader_k16.csv")[labels]
    right = pd.read_csv("/kaggle/working/p006_reader_k24.csv")[labels]
    comparison = {
        "max_probability_delta": float((right - left).abs().to_numpy().max()),
        "mean_absolute_probability_delta": float((right - left).abs().to_numpy().mean()),
        "per_target_rank_correlation": {label: float(left[label].rank().corr(right[label].rank())) for label in labels},
        "accuracy_evaluated": False,
        "note": "Label-free differences are not evidence of AUC improvement; public checkpoints are fullfit.",
    }
    # A diagnostic notebook deliberately does not provide the competition output filename.
    Path("/kaggle/working/submission.csv").unlink()
    return reports, comparison
```

```python
%%writefile /kaggle/working/p010_src/p007_metrics.py
"""Paired ConvNeXt plain/mirror observations, embedded only in a diagnostic Notebook."""

import ast
import contextlib
import io
import json
import math
import shutil
import sys
from pathlib import Path

from p006_metrics import Resources, check_csv, reader_observer, sha256, write_json


@contextlib.contextmanager
def inference_observer(namespace, mode, output_root="/kaggle/working"):
    """Record actual top-level forwards and TTA calls without replacing their outputs."""
    if mode not in ("plain", "mirror"):
        raise ValueError("Expected plain or mirror observation mode")
    directory = Path(output_root) / f"p007_observation_{mode}"
    directory.mkdir(exist_ok=False)
    state = {"mode": mode, "model_count": None, "load_grad_disabled": None,
             "forwards": [], "mirror_calls": [], "complete": False}
    original_load = namespace["load_models"]
    original_mirror = namespace.get("anatomical_lr_mirror")
    if (mode == "mirror") != (original_mirror is not None):
        raise ValueError("Mirror helper presence does not match the expected source")
    handles = []

    def observed_load(*args, **kwargs):
        import torch

        assert not torch.is_grad_enabled(), "Production run must disable gradients before loading models"
        state["load_grad_disabled"] = True
        models = original_load(*args, **kwargs)
        assert state["model_count"] is None, "Model loading repeated unexpectedly"
        state["model_count"] = len(models)
        assert len(models) == 3, "The pinned reader requires the same three checkpoints"
        for model_index, (model, _) in enumerate(models):
            assert all(not part.training for part in model.modules())

            def observe_forward(module, inputs, output, model_index=model_index):
                import torch

                assert not torch.is_grad_enabled(), "Production reader forward enabled gradients"
                x = inputs[0]
                total = len(state["forwards"])
                phase = "plain" if mode == "plain" or total % 6 < 3 else "mirror"
                assert model_index == total % 3, "Original model mean order changed"
                assert list(x.shape[1:]) == [6, 16, 3, 384, 384]
                assert output.shape == (len(x), 12) and bool(torch.isfinite(output).all())
                state["forwards"].append({
                    "model_index": model_index, "phase": phase, "shape": list(x.shape),
                    "finite_logits": True, "grad_disabled": True,
                })
                # None preserves the exact original output object.

            handles.append(model.register_forward_hook(observe_forward))
        return models

    def observed_mirror(x, mask, pos):
        import torch

        assert len(state["forwards"]) % 6 == 3, "Mirror must follow the three plain forwards"
        result = original_mirror(x, mask, pos)
        xm, maskm, posm = result
        assert maskm is mask and xm.shape == x.shape and xm.dtype == x.dtype == torch.uint8
        assert posm.shape == pos.shape and posm.dtype == pos.dtype
        for slot in (0, 3):
            expected = torch.where(mask[:, slot, None], 1.0 - pos[:, slot], pos[:, slot])
            assert torch.equal(posm[:, slot], expected)
        for slot in (1, 2, 4, 5):
            assert torch.equal(posm[:, slot], pos[:, slot])
        state["mirror_calls"].append({
            "shape": list(x.shape), "valid_slots": mask.tolist(), "mask_object_preserved": True,
            "position_contract_passed": True, "uint8_before_normalization": True,
        })
        return result

    namespace["load_models"] = observed_load
    if mode == "mirror":
        namespace["anatomical_lr_mirror"] = observed_mirror
    try:
        with reader_observer(namespace["DicomStudyDataset"], 16, directory):
            yield state
        state["complete"] = True
    finally:
        namespace["load_models"] = original_load
        if mode == "mirror":
            namespace["anatomical_lr_mirror"] = original_mirror
        for handle in handles:
            handle.remove()
        write_json(directory / "inference.json", state)


def observed_reader_source(source, mode):
    origin = "    run(data_dir, split, ckpts).to_csv(out_csv)"
    assert source.count(origin) == 1
    assert source.count("def run(data_dir, split, ckpts, k=16, studies=None, workers=4, bs=4):") == 1
    tree = ast.parse(source)
    runs = [node for node in tree.body if isinstance(node, ast.FunctionDef) and node.name == "run"]
    assert len(runs) == 1 and [ast.unparse(node) for node in runs[0].decorator_list] == ["torch.no_grad()"], (
        "Original production run no_grad decorator must stay attached to run"
    )
    invocation = (
        "    from p007_metrics import inference_observer\n"
        f"    with inference_observer(globals(), {mode!r}):\n"
        "        run(data_dir, split, ckpts).to_csv(out_csv)"
    )
    return source.replace(origin, invocation, 1)


def validate_observations(state, mode, studies):
    batches = math.ceil(studies / 4)
    views = 1 if mode == "plain" else 2
    assert state["mode"] == mode and state["complete"]
    assert state["model_count"] == 3
    assert state["load_grad_disabled"] is True
    assert len(state["forwards"]) == batches * 3 * views
    assert len(state["mirror_calls"]) == (0 if mode == "plain" else batches)
    assert all(row["finite_logits"] and row["grad_disabled"] for row in state["forwards"])
    assert sum(row["shape"][0] for row in state["forwards"] if row["model_index"] == 0
               and row["phase"] == "plain") == studies
    if mode == "mirror":
        assert sum(row["shape"][0] for row in state["forwards"] if row["model_index"] == 0
                   and row["phase"] == "mirror") == studies
        assert all(row["position_contract_passed"] and row["mask_object_preserved"]
                   for row in state["mirror_calls"])
    return {"batches": batches, "views": views, "model_forward_calls": len(state["forwards"]),
            "mirror_transform_calls": len(state["mirror_calls"]), "grad_disabled_at_load_and_all_forwards": True,
            "all_passed": True}


def paired_mirror_readers(namespace, fusion_source, plain_source, mirror_source, uids, labels):
    """Reuse one CoAt CSV; execute plain once and the exact p007 two-view reader once."""
    import numpy as np
    import pandas as pd
    import torch

    reports = {}
    for mode, reader_source in (("plain", plain_source), ("mirror", mirror_source)):
        shutil.copyfile("/kaggle/working/_sota_0949.csv", "/kaggle/working/submission.csv")
        Path("/kaggle/working/_own.csv").unlink(missing_ok=True)
        runtime_source = observed_reader_source(reader_source, mode)
        path = Path("/kaggle/working/own_src/infer.py")
        path.write_text(runtime_source, encoding="utf-8")
        captured, destination = io.StringIO(), sys.stdout

        class Tee:
            def write(self, text):
                destination.write(text)
                return captured.write(text)

            def flush(self):
                destination.flush()

        with Resources() as resources, contextlib.redirect_stdout(Tee()):
            exec(compile(fusion_source, f"p007-fusion-{mode}", "exec"), namespace)
        log = captured.getvalue()
        assert log.count("[Apex Fusion SUCCESS] Successfully fused") == 1
        assert "[Apex Fusion EXCEPTION]" not in log and "[Apex Fusion FALLBACK]" not in log
        own, fused = Path("/kaggle/working/_own.csv"), Path("/kaggle/working/submission.csv")
        own_record, fused_record = check_csv(own, uids, labels), check_csv(fused, uids, labels)
        expected = pd.read_csv("/kaggle/working/_sota_0949.csv")[labels].rank(pct=True)
        reader = pd.read_csv(own)[labels].rank(pct=True)
        for label in labels:
            weight = namespace["TARGET_WEIGHTS"][label]
            expected[label] = (1.0 - weight) * expected[label] + weight * reader[label]
        expected = expected.rank(pct=True)
        assert np.allclose(pd.read_csv(fused)[labels], expected, atol=1e-12, rtol=0)
        directory = Path(f"/kaggle/working/p007_observation_{mode}")
        coverage = [json.loads(p.read_text(encoding="utf-8"))
                    for p in sorted((directory / "p006_cov_k16").glob("*.json"))]
        assert [row["uid"] for row in coverage] == uids
        assert all(row["shape"] == [6, 16, 3, 384, 384] for row in coverage)
        assert all(any(row["mask"]) for row in coverage)
        state = json.loads((directory / "inference.json").read_text(encoding="utf-8"))
        observed = validate_observations(state, mode, len(uids))
        reports[mode] = {
            **resources.result(), "reader_csv": own_record, "fused_csv": fused_record,
            "coverage": coverage, "inference_observations": state, "forward_gate": observed,
            "observed_runtime_reader_sha256": sha256(path),
            "fusion_matches_fixed_formula": True, "fusion_success_marker": True,
            "fusion_exception_or_fallback": False,
            "observation_note": "Top-level finite-logit hooks synchronize CUDA; elapsed includes observers, pip and cache effects. This is a feasibility profile, not a pure model speed comparison.",
        }
        shutil.copyfile(own, f"/kaggle/working/p007_reader_{mode}.csv")
        shutil.copyfile(fused, f"/kaggle/working/p007_fused_{mode}.csv")
        torch.cuda.empty_cache()
    assert [r["mask"] for r in reports["plain"]["coverage"]] == [
        r["mask"] for r in reports["mirror"]["coverage"]
    ]
    left = pd.read_csv("/kaggle/working/p007_reader_plain.csv")[labels]
    right = pd.read_csv("/kaggle/working/p007_reader_mirror.csv")[labels]
    comparison = {
        "max_probability_delta": float((right - left).abs().to_numpy().max()),
        "mean_absolute_probability_delta": float((right - left).abs().to_numpy().mean()),
        "per_target_rank_correlation": {
            label: float(left[label].rank().corr(right[label].rank())) for label in labels
        },
        "k_fixed": 16, "main_coat_executions": 1, "accuracy_evaluated": False,
        "note": "Differences and correlations do not show AUC improvement; public models are fullfit.",
    }
    Path("/kaggle/working/submission.csv").unlink()
    return reports, comparison
```

```python
%%writefile /kaggle/working/p010_src/p009_metrics.py
"""Label-free paired CoAt FOV observations; one unchanged ConvNeXt reader execution."""

import ast
import base64
import contextlib
import csv
import hashlib
import importlib.util
import io
import json
import math
import shutil
import subprocess
import sys
import types
from pathlib import Path

from p006_metrics import Resources, check_csv, sha256
from p007_metrics import observed_reader_source, validate_observations


def embedded_source(main_source):
    nodes = [node for node in ast.parse(main_source).body if isinstance(node, ast.Assign)
             and len(node.targets) == 1 and isinstance(node.targets[0], ast.Name)
             and node.targets[0].id == "_FASTREAD_B64"]
    assert len(nodes) == 1
    return base64.b64decode(ast.literal_eval(nodes[0].value), validate=True)


def prepare_fastread(main_source, allowed_source_hashes, working="/kaggle/working"):
    """Overwrite only this diagnostic source; remove only its exact import-cache file."""
    root = Path(working).resolve()
    path = root / "_fastread.py"
    assert path.resolve().parent == root, "Fastread source must remain within the exact working root"
    payload = embedded_source(main_source)
    expected_hash = hashlib.sha256(payload).hexdigest()
    assert expected_hash in allowed_source_hashes
    old_hash = sha256(path) if path.exists() else None
    assert old_hash is None or old_hash in allowed_source_hashes, "Unexpected pre-existing fastread source"
    cache = Path(importlib.util.cache_from_source(str(path)))
    assert cache.resolve().parent.parent == root and cache.parent.name == "__pycache__", (
        "Exact fastread cache path must remain within this diagnostic working root"
    )
    cache_deleted = cache.is_file()
    if cache_deleted:
        cache.unlink()
    path.write_bytes(payload)
    assert sha256(path) == expected_hash
    return {"source_sha256": expected_hash, "previous_source_sha256": old_hash,
            "specific_cache_file": str(cache), "specific_cache_deleted": cache_deleted,
            "no_other_cache_or_source_removed": True}


@contextlib.contextmanager
def registered_namespace(name, values=None):
    """Expose exec-defined functions to the module lookup used by Pool task pickling."""
    assert name.isidentifier() and name != "__main__" and name not in sys.modules
    module = types.ModuleType(name)
    if values:
        assert "__name__" not in values
        module.__dict__.update(values)
    sys.modules[name] = module
    try:
        yield module.__dict__
    finally:
        assert sys.modules.get(name) is module, "Unexpected replacement of the diagnostic module"
        del sys.modules[name]


def fork_transport_probe():
    """Pure artificial task on the real Linux fork context, before any model/MRI processing."""
    import multiprocessing as mp
    import pickle
    from multiprocessing.reduction import ForkingPickler

    assert "fork" in mp.get_all_start_methods(), "The unchanged original main requires Linux fork"
    name = "_p009_artificial_fork_probe"
    with registered_namespace(name) as namespace:
        exec("def task(value):\n    return value + 3\n", namespace)
        task = namespace["task"]
        assert pickle.loads(ForkingPickler.dumps(task)) is task
        with mp.get_context("fork").Pool(1) as pool:
            actual = pool.map(task, [1, 2, 3])
        assert actual == [4, 5, 6]
    assert name not in sys.modules
    return {"context": "fork", "workers": 1, "artificial_tasks": 3,
            "task_roundtrip_passed": True, "registered_module_removed": True,
            "MRI_model_labels_or_GPU_consumed": False}


def observed_main_source(main_source, root="/kaggle/working/p009_profile_data"):
    """Original predictions plus returned masks, worker binding and forward observers."""
    marker = "    ROOT = find_test_root()"
    assert main_source.count(marker) == 1
    source = main_source.replace(marker, f"    ROOT = {root!r}", 1)
    marker = "            q.put((sid, vol, mask))"
    assert source.count(marker) == 1
    source = source.replace(marker, "            _p009_masks[sid] = int(mask.sum())\n" + marker, 1)
    tail = 'if __name__ == "__main__":\n    main()'
    assert source.count(tail) == 1
    return source.replace(tail, '''_p009_masks = {}
_p009_pairs = []
_p009_forwards = []
_p009_loaded_devices = []
_p009_hooks = []
_p009_original_pair = infer_pair_probs
_p009_original_init = _pre_init
_p009_original_load = load_model

def _p009_observe_init():
    _p009_original_init()
    import hashlib as _h, json as _j
    assert float(_READER.__globals__["CROP_MM"]) == _p009_expected_fov
    _p009_file = _READER.__globals__["__file__"]
    with open(_p009_file, "rb") as _f:
        _p009_hash = _h.file_digest(_f, "sha256").hexdigest()
    assert _p009_hash == _p009_expected_source_hash
    _p009_closure = dict(zip(_READER.__code__.co_freevars,
                              [cell.cell_contents for cell in _READER.__closure__], strict=True))
    assert _p009_closure["IMG"] == IMG == 384 and _p009_closure["SLOTS"] == SLOTS
    _p009_record = {"pid":os.getpid(), "actual_builder_CROP_MM":float(_READER.__globals__["CROP_MM"]),
                    "actual_builder_IMG":int(_p009_closure["IMG"]),
                    "actual_builder_SLOTS":_p009_closure["SLOTS"],
                    "actual_builder_source_sha256":_p009_hash}
    with open(os.path.join(_p009_worker_directory, str(os.getpid())+".json"), "w") as _f:
        _j.dump(_p009_record, _f)

def _p009_observe_load(*args, **kwargs):
    model, res = _p009_original_load(*args, **kwargs)
    assert res == 320 and all(not module.training for module in model.modules())
    _p009_loaded_devices.append(str(next(model.parameters()).device))
    def _p009_forward(module, inputs, output):
        assert not torch.is_grad_enabled()
        assert list(inputs[0].shape) == [1,94,3,320,320]
        assert list(output.shape) == [1,12] and bool(torch.isfinite(output).all())
        _p009_forwards.append({"shape":list(inputs[0].shape),"device":str(inputs[0].device),
                               "grad_disabled":True,"finite_logits":True})
    _p009_hooks.append(model.register_forward_hook(_p009_forward))
    return model, res

def _p009_observe_pair(model, plain, mirrored, device):
    assert plain.shape == mirrored.shape == (94,3,320,320)
    result = _p009_original_pair(model, plain, mirrored, device)
    assert bool(np.isfinite(result).all()) and result.shape == (12,)
    _p009_pairs.append({"device":str(device),"shape":[94,3,320,320],"finite":True})
    return result

_pre_init = _p009_observe_init
load_model = _p009_observe_load
infer_pair_probs = _p009_observe_pair
import pickle as _p009_pickle
from multiprocessing.reduction import ForkingPickler as _p009_ForkingPickler
_p009_task_pickle_identity = {}
for _p009_task_name in ("_pre_one", "_pre_init"):
    _p009_task = globals()[_p009_task_name]
    assert _p009_pickle.loads(_p009_ForkingPickler.dumps(_p009_task)) is _p009_task
    _p009_task_pickle_identity[_p009_task_name] = True
main()
for _p009_hook in _p009_hooks:
    _p009_hook.remove()
''', 1)


@contextlib.contextmanager
def capture_stdout():
    captured, destination = io.StringIO(), sys.stdout

    class Tee:
        def write(self, value):
            destination.write(value)
            return captured.write(value)

        def flush(self):
            destination.flush()

    with contextlib.redirect_stdout(Tee()):
        yield captured


def run_main(main_source, fov, uids, labels, allowed_hashes):
    binding = prepare_fastread(main_source, allowed_hashes)
    worker_directory = Path(f"/kaggle/working/p009_workers_fov{fov}")
    worker_directory.mkdir(exist_ok=False)
    values = {"__file__": f"/kaggle/working/p009_main_fov{fov}.py",
                 "_p009_expected_fov": float(fov),
                 "_p009_expected_source_hash": binding["source_sha256"],
                 "_p009_worker_directory": str(worker_directory)}
    for name in ("submission.csv", "_sota_0949.csv"):
        Path("/kaggle/working", name).unlink(missing_ok=True)
    observed_source = observed_main_source(main_source)
    module_name = f"_p009_profile_main_fov{fov}"
    with registered_namespace(module_name, values) as namespace:
        with Resources() as resources, capture_stdout() as captured:
            exec(compile(observed_source, f"p009-main-fov{fov}", "exec"), namespace)
    assert module_name not in sys.modules
    assert namespace["_p009_task_pickle_identity"] == {"_pre_one": True, "_pre_init": True}
    assert captured.getvalue().count("wrote /kaggle/working/submission.csv |") == 1
    assert "FALLBACK" not in captured.getvalue() and "FAILED (" not in captured.getvalue()
    assert set(namespace["_p009_masks"]) == set(uids)
    assert all(value > 0 for value in namespace["_p009_masks"].values())
    assert len(namespace["ARMS"]) == 1
    assert len(namespace["_p009_pairs"]) == len(uids)
    assert len(namespace["_p009_forwards"]) == len(uids) * 2
    assert len(namespace["_p009_loaded_devices"]) == namespace["torch"].cuda.device_count()
    workers = [json.loads(path.read_text(encoding="utf-8")) for path in sorted(worker_directory.glob("*.json"))]
    assert workers and len({row["pid"] for row in workers}) == len(workers)
    assert all(row["actual_builder_CROP_MM"] == fov and row["actual_builder_IMG"] == 384
               and row["actual_builder_source_sha256"] == binding["source_sha256"] for row in workers)
    destination = Path(f"/kaggle/working/p009_main_fov{fov}.csv")
    shutil.copyfile("/kaggle/working/_sota_0949.csv", destination)
    record = check_csv(destination, uids, labels)
    namespace["torch"].cuda.empty_cache()
    return {
        **resources.result(), "fov_mm": fov, "main_csv": record, "fastread_binding": binding,
        "worker_initializers": workers, "valid_slices_by_uid": namespace["_p009_masks"],
        "successful_pair_calls": len(namespace["_p009_pairs"]),
        "forward_observations": namespace["_p009_forwards"],
        "loaded_model_devices": namespace["_p009_loaded_devices"],
        "main_success_marker": True, "main_fallback_or_preprocess_failure": False,
        "prediction_source_sha256": hashlib.sha256(main_source.encode()).hexdigest(),
        "observed_runtime_source_sha256": hashlib.sha256(observed_source.encode()).hexdigest(),
        "task_pickle_identity": namespace["_p009_task_pickle_identity"],
        "registered_module_name": module_name, "registered_module_removed": True,
    }


@contextlib.contextmanager
def replay_reader_once(cache, expected_sha, root="/kaggle/working"):
    """Replay only the exact reader subprocess on the second fusion; restore on any failure."""
    cache = Path(cache)
    assert sha256(cache) == expected_sha
    reader = str(Path(root) / "own_src/infer.py")
    output = str(Path(root) / "_own.csv")
    alias = str(Path(root) / "p009_profile_data")
    original = subprocess.run
    state = {"replay_calls": 0}

    def replay(command, *args, **kwargs):
        if isinstance(command, (list, tuple)) and len(command) > 1 and command[1] == reader:
            assert len(command) == 8 and command[2:5] == [alias, "test", output]
            assert kwargs.get("check") is True and "env" in kwargs
            assert state["replay_calls"] == 0 and sha256(cache) == expected_sha
            assert not Path(output).exists(), "Old reader CSV must be removed before replay"
            shutil.copyfile(cache, output)
            state["replay_calls"] += 1
            return subprocess.CompletedProcess(command, 0)
        return original(command, *args, **kwargs)

    subprocess.run = replay
    try:
        yield state
    finally:
        subprocess.run = original


def percentile_rank(values):
    indices = sorted(range(len(values)), key=values.__getitem__)
    output = [0.] * len(values)
    start = 0
    while start < len(indices):
        stop = start + 1
        while stop < len(indices) and values[indices[stop]] == values[indices[start]]:
            stop += 1
        value = ((start + 1 + stop) / 2) / len(values)
        for index in indices[start:stop]:
            output[index] = value
        start = stop
    return output


def fusion_maxdiff(main_path, reader_path, fused_path, uids, labels, weights):
    """Independent standard-library average-tie percentile rank formula."""
    records = []
    for path in (main_path, reader_path, fused_path):
        with Path(path).open(encoding="utf-8", newline="") as stream:
            rows = list(csv.DictReader(stream))
        assert [row["StudyInstanceUID"] for row in rows] == uids
        records.append(rows)
    maximum = 0.
    for label in labels:
        main = percentile_rank([float(row[label]) for row in records[0]])
        reader = percentile_rank([float(row[label]) for row in records[1]])
        weight = weights[label]
        expected = percentile_rank([(1 - weight) * a + weight * b for a, b in zip(main, reader, strict=True)])
        actual = [float(row[label]) for row in records[2]]
        maximum = max(maximum, max(abs(a - b) for a, b in zip(expected, actual, strict=True)))
    assert maximum <= 1e-12
    return maximum


def run_fusion(namespace, fusion_source, fov, uids, labels, reader_cache, replay=False):
    root = Path("/kaggle/working")
    main = root / f"p009_main_fov{fov}.csv"
    shutil.copyfile(main, root / "_sota_0949.csv")
    shutil.copyfile(main, root / "submission.csv")
    (root / "_own.csv").unlink(missing_ok=True)
    context = replay_reader_once(reader_cache, sha256(reader_cache)) if replay else contextlib.nullcontext()
    with context as replay_state, Resources() as resources, capture_stdout() as captured:
        exec(compile(fusion_source, f"p009-fusion-fov{fov}", "exec"), namespace)
    log = captured.getvalue()
    assert log.count("[Apex Fusion SUCCESS] Successfully fused") == 1
    assert "[Apex Fusion EXCEPTION]" not in log and "[Apex Fusion FALLBACK]" not in log
    own, fused = root / "_own.csv", root / "submission.csv"
    own_record, fused_record = check_csv(own, uids, labels), check_csv(fused, uids, labels)
    if replay:
        assert replay_state["replay_calls"] == 1 and own_record["sha256"] == sha256(reader_cache)
    else:
        assert not reader_cache.exists()
        shutil.copyfile(own, reader_cache)
    difference = fusion_maxdiff(main, reader_cache, fused, uids, labels, namespace["TARGET_WEIGHTS"])
    shutil.copyfile(fused, root / f"p009_fused_fov{fov}.csv")
    return {**resources.result(), "reader_csv": own_record, "fused_csv": fused_record,
            "fusion_max_difference_fixed_formula": difference, "fusion_success_marker": True,
            "fusion_exception_or_fallback": False, "reader_replayed": replay,
            "reader_replay_calls": replay_state["replay_calls"] if replay else 0}


def paired_fov(namespace, original_main, candidate_main, fusion_source, infer_source, uids, labels):
    import numpy as np
    import pandas as pd
    import torch

    hashes = {hashlib.sha256(embedded_source(value)).hexdigest() for value in (original_main, candidate_main)}
    assert len(hashes) == 2 and len(uids) == 32
    transport = fork_transport_probe()
    runtime_reader = observed_reader_source(infer_source, "plain")
    reader_path = Path("/kaggle/working/own_src/infer.py")
    reader_path.write_text(runtime_reader, encoding="utf-8")
    reader_cache = Path("/kaggle/working/p009_reader_k16.csv")
    assert not reader_cache.exists()
    main_reports, fused_reports = {}, {}
    for fov, main_source in ((140, original_main), (160, candidate_main)):
        main_reports[str(fov)] = run_main(main_source, fov, uids, labels, hashes)
        fused_reports[str(fov)] = run_fusion(namespace, fusion_source, fov, uids, labels, reader_cache, replay=fov == 160)
    directory = Path("/kaggle/working/p007_observation_plain")
    coverage = [json.loads(path.read_text(encoding="utf-8"))
                for path in sorted((directory / "p006_cov_k16").glob("*.json"))]
    assert [row["uid"] for row in coverage] == uids
    assert all(row["shape"] == [6, 16, 3, 384, 384] and any(row["mask"]) for row in coverage)
    state = json.loads((directory / "inference.json").read_text(encoding="utf-8"))
    gate = validate_observations(state, "plain", len(uids))
    assert fused_reports["140"]["reader_csv"]["sha256"] == fused_reports["160"]["reader_csv"]["sha256"]
    left = pd.read_csv("/kaggle/working/p009_main_fov140.csv")[labels]
    right = pd.read_csv("/kaggle/working/p009_main_fov160.csv")[labels]
    correlations = {label: float(left[label].rank().corr(right[label].rank())) for label in labels}
    assert all(math.isfinite(value) for value in correlations.values())
    comparison = {
        "max_main_rank_delta": float(np.abs(right.to_numpy() - left.to_numpy()).max()),
        "mean_absolute_main_rank_delta": float(np.abs(right.to_numpy() - left.to_numpy()).mean()),
        "per_target_main_rank_correlation": correlations,
        "valid_slice_count_differences": {uid: main_reports["160"]["valid_slices_by_uid"][uid]
                                         - main_reports["140"]["valid_slices_by_uid"][uid] for uid in uids},
        "main_coat_executions": 2, "reader_convnext_executions": 1, "reader_k_fixed": 16,
        "fusion_executions": 2, "second_fusion_reader_csv_replayed_once": True,
        "accuracy_evaluated": False,
        "note": "Label-free prediction differences do not establish AUC improvement. Public weights are fullfit.",
    }
    Path("/kaggle/working/submission.csv").unlink()
    assert not Path("/kaggle/working/submission.csv").exists()
    torch.cuda.empty_cache()
    return {
        "main": main_reports, "fusion": fused_reports,
        "reader": {"coverage": coverage, "inference_observations": state, "forward_gate": gate,
                   "runtime_reader_sha256": sha256(reader_path), "reader_csv": check_csv(reader_cache, uids, labels)},
        "comparison": comparison, "final_submission_csv_absent": True,
        "artificial_fork_transport": transport,
    }
```

```python
%%writefile /kaggle/working/p010_src/p010_shared_metrics.py
"""Bounded transport and fixed rank blending for the unqualified p010 candidate.

This module never downloads, decodes, trains, uploads or submits anything.
Transport is a separately controlled execution change, not an accuracy claim.
"""

import csv
import hashlib
import io
import math
import os
import queue
import sys
import threading
import time
from pathlib import Path


class PipelineFailure(RuntimeError):
    """Preparation or inference failed; no partial predictions may be submitted."""


class FusionObserver:
    """Transparent stdout provenance guard around the literal original fusion."""

    def __init__(self):
        self.original = sys.stdout
        self.buffer = io.StringIO()
        self.restored = False
        sys.stdout = self

    def write(self, value):
        self.buffer.write(value)
        return self.original.write(value)

    def flush(self):
        return self.original.flush()

    def __getattr__(self, name):
        return getattr(self.original, name)

    def restore(self):
        if self.restored or sys.stdout is not self:
            raise PipelineFailure("Fusion stdout observer identity or lifetime changed")
        sys.stdout = self.original
        self.restored = True
        return self.buffer.getvalue()

    def require_success(self):
        value = self.restore()
        counts = {
            "success": value.count("[Apex Fusion SUCCESS] Successfully fused"),
            "exception": value.count("[Apex Fusion EXCEPTION]"),
            "fallback": value.count("[Apex Fusion FALLBACK]"),
        }
        if counts != {"success": 1, "exception": 0, "fallback": 0}:
            raise PipelineFailure(f"Original incumbent fusion did not finish exactly once: {counts}")
        return {
            "original_fusion_stdout_guard": counts,
            "original_fusion_success_verified": True,
            "counts_from_parent_inprocess_stdout_not_frontend_repeated_display": True,
        }


def run_bounded_pipeline(
    pool_factory,
    task,
    items,
    devices,
    consume,
    *,
    queue_max=48,
    outstanding_limit=48,
    shutdown_seconds=120.0,
    poll_seconds=0.05,
):
    """Deliver completed prep tasks without Pool.imap's unbounded result buffer.

    At most outstanding_limit task results are pending delivery, at most
    queue_max are queued for inference, and one is owned by each consumer.
    Callbacks never block the Pool result handler. A new task is submitted only
    after a previous result has been accepted by the inference queue.
    """
    if not devices or queue_max < 1 or outstanding_limit < 1 or shutdown_seconds <= 0 or poll_seconds <= 0:
        raise ValueError("Positive capacities, shutdown time and devices are required")
    values = list(items)
    if len(values) != len(set(values)):
        raise ValueError("Task identities must be unique")
    delivery = queue.Queue(maxsize=outstanding_limit)
    gpu_queue = queue.Queue(maxsize=queue_max)
    cancel = threading.Event()
    lock = threading.Lock()
    first_error = []
    counters = {
        "submitted": 0,
        "delivered": 0,
        "consumed": 0,
        "discarded_after_cancel": 0,
        "max_outstanding": 0,
        "max_delivery_queue": 0,
        "max_inference_queue": 0,
    }

    def fail(origin, exception):
        with lock:
            if not first_error:
                first_error.append((origin, type(exception).__name__, str(exception)))
        cancel.set()

    def callback(value):
        try:
            delivery.put_nowait((True, value))
            with lock:
                counters["max_delivery_queue"] = max(counters["max_delivery_queue"], delivery.qsize())
        except BaseException as exception:
            fail("result_callback", exception)

    def error_callback(exception):
        fail("preparation_worker", exception)

    def consumer(device):
        while True:
            item = gpu_queue.get()
            try:
                if item is None:
                    return
                if cancel.is_set():
                    with lock:
                        counters["discarded_after_cancel"] += 1
                    continue
                try:
                    consume(device, item)
                    with lock:
                        counters["consumed"] += 1
                except BaseException as exception:
                    fail(f"inference_consumer:{device}", exception)
            finally:
                gpu_queue.task_done()
                del item

    threads = [threading.Thread(target=consumer, args=(device,), daemon=True) for device in devices]
    for thread in threads:
        thread.start()
    pool = None
    exceptional = None
    next_index = 0
    outstanding = 0
    try:
        pool = pool_factory()

        def submit_one():
            nonlocal next_index, outstanding
            value = values[next_index]
            pool.apply_async(task, (value,), callback=callback, error_callback=error_callback)
            next_index += 1
            outstanding += 1
            with lock:
                counters["submitted"] += 1
                counters["max_outstanding"] = max(counters["max_outstanding"], outstanding)

        while next_index < len(values) and outstanding < outstanding_limit:
            submit_one()
        while outstanding:
            if cancel.is_set():
                raise PipelineFailure("Worker or inference consumer failed")
            try:
                succeeded, payload = delivery.get(timeout=poll_seconds)
            except queue.Empty:
                continue
            assert succeeded
            try:
                accepted = False
                while not accepted:
                    if cancel.is_set():
                        raise PipelineFailure("Worker or inference consumer failed")
                    try:
                        gpu_queue.put(payload, timeout=poll_seconds)
                        accepted = True
                        with lock:
                            counters["max_inference_queue"] = max(counters["max_inference_queue"], gpu_queue.qsize())
                            counters["delivered"] += 1
                    except queue.Full:
                        continue
            finally:
                del payload
                delivery.task_done()
            outstanding -= 1
            if next_index < len(values):
                submit_one()
        pool.close()
        pool.join()
    except BaseException as exception:
        exceptional = exception
        fail("producer", exception)
        if pool is not None:
            pool.terminate()
            pool.join()
    finally:
        deadline = time.monotonic() + shutdown_seconds
        for _ in threads:
            while True:
                if time.monotonic() >= deadline:
                    raise PipelineFailure("Inference consumers did not drain before shutdown deadline")
                try:
                    gpu_queue.put(None, timeout=poll_seconds)
                    break
                except queue.Full:
                    continue
        for thread in threads:
            thread.join(max(0.0, deadline - time.monotonic()))
        if any(thread.is_alive() for thread in threads):
            raise PipelineFailure("Inference consumers did not stop before shutdown deadline")
        if gpu_queue.unfinished_tasks != 0:
            raise PipelineFailure("Inference queue task_done contract failed")
        while True:
            try:
                delivery.get_nowait()
                delivery.task_done()
            except queue.Empty:
                break
        if delivery.unfinished_tasks != 0:
            raise PipelineFailure("Delivery queue task_done contract failed")
    receipt = {
        "schema_version": "p010_bounded_transport_v1",
        "tasks": len(values),
        "queue_max": queue_max,
        "outstanding_limit": outstanding_limit,
        "consumer_count": len(devices),
        "prepared_parent_payload_upper_bound": queue_max + outstanding_limit + len(devices) + 1,
        "pool_result_handler_last_payload_reserve": 1,
        "bound_excludes_worker_and_IPC_serialisation_memory": True,
        "all_tasks_consumed": counters["consumed"] == len(values),
        "queue_unfinished_tasks": gpu_queue.unfinished_tasks,
        "delivery_unfinished_tasks": delivery.unfinished_tasks,
        "errors": len(first_error),
        "cancelled": cancel.is_set(),
        "consumer_threads_joined": True,
        "delivery": "bounded_apply_async_single_study_tasks",
        **counters,
    }
    if first_error:
        origin, kind, message = first_error[0]
        failure = PipelineFailure(f"{origin}: {kind}: {message}")
        failure.receipt = receipt
        raise failure from exceptional
    assert counters["submitted"] == counters["delivered"] == counters["consumed"] == len(values)
    assert counters["discarded_after_cancel"] == 0
    return receipt


def average_percentile(values):
    """Average ties, ranks 1..N divided by N, as pandas rank(pct=True)."""
    if not values or any(not math.isfinite(value) for value in values):
        raise ValueError("Finite, nonempty probabilities are required")
    order = sorted(range(len(values)), key=values.__getitem__)
    result = [0.0] * len(values)
    left = 0
    while left < len(order):
        right = left + 1
        while right < len(order) and values[order[right]] == values[order[left]]:
            right += 1
        rank = ((left + 1) + right) / (2 * len(order))
        for position in order[left:right]:
            result[position] = rank
        left = right
    return result


def shared_input_receipt(volume, mask):
    """Hash exactly the shared native uint8 bytes; never cast, resize or decode."""
    import numpy as np

    if volume.shape != (96, 384, 384) or volume.dtype != np.uint8 or not volume.flags.c_contiguous:
        raise ValueError("Original contiguous 96x384x384 uint8 volume is required")
    if mask.shape != (96,) or mask.dtype != np.uint8 or not mask.flags.c_contiguous:
        raise ValueError("Original contiguous 96-entry uint8 mask is required")
    if not bool(np.isin(mask, [0, 1]).all()):
        raise ValueError("Mask must retain the original binary slot contract")
    return {
        "volume_sha256": hashlib.sha256(memoryview(volume).cast("B")).hexdigest(),
        "mask_sha256": hashlib.sha256(memoryview(mask).cast("B")).hexdigest(),
        "volume_shape": list(volume.shape),
        "volume_dtype": str(volume.dtype),
        "mask_shape": list(mask.shape),
        "mask_dtype": str(mask.dtype),
        "valid_slices": int(mask.sum()),
        "volume_bytes": int(volume.nbytes),
        "mask_bytes": int(mask.nbytes),
    }


def read_submission(path, labels, expected_uids=None):
    with Path(path).open(encoding="utf-8", newline="") as handle:
        reader = csv.DictReader(handle)
        if reader.fieldnames != ["StudyInstanceUID", *labels] or len(labels) != 12:
            raise ValueError("Exactly twelve findings in the original label order are required")
        rows = list(reader)
        if any(set(row) != set(reader.fieldnames) or any(value is None for value in row.values()) for row in rows):
            raise ValueError("Every row must retain the exact thirteen-column schema")
    uids = [row["StudyInstanceUID"] for row in rows]
    if not uids or any(not uid for uid in uids) or len(set(uids)) != len(uids):
        raise ValueError("Nonempty unique study identities are required")
    if expected_uids is not None and uids != list(expected_uids):
        raise ValueError("Study order or identity changed")
    matrix = [[float(row[label]) for label in labels] for row in rows]
    if any(not math.isfinite(value) or not 0 <= value <= 1 for row in matrix for value in row):
        raise ValueError("Finite probabilities in [0,1] are required")
    return uids, matrix


def write_submission_exclusive(path, uids, matrix, labels):
    path = Path(path)
    if len(matrix) != len(uids) or any(len(row) != 12 for row in matrix):
        raise ValueError("Submission matrix shape changed")
    with path.open("x", encoding="utf-8", newline="") as handle:
        writer = csv.writer(handle, lineterminator="\n")
        writer.writerow(["StudyInstanceUID", *labels])
        for uid, row in zip(uids, matrix, strict=True):
            writer.writerow([uid, *[format(float(value), ".17g") for value in row]])
    read_submission(path, labels, uids)


def fixed_rank10(incumbent, expert):
    if len(incumbent) != len(expert) or not incumbent:
        raise ValueError("Branches must cover the same nonempty studies")
    if any(len(row) != 12 for matrix in (incumbent, expert) for row in matrix):
        raise ValueError("Both branches must retain all twelve original findings")
    columns = []
    for label in range(12):
        left = average_percentile([row[label] for row in incumbent])
        right = average_percentile([row[label] for row in expert])
        columns.append(average_percentile([0.9 * a + 0.1 * b for a, b in zip(left, right, strict=True)]))
    return [[column[index] for column in columns] for index in range(len(incumbent))]


def finalise_fixed_rank10(incumbent_path, expert_path, labels, uids, *, working=None):
    """Retain exact incumbent bytes and atomically replace a validated candidate."""
    incumbent_path, expert_path = Path(incumbent_path), Path(expert_path)
    root = Path(working) if working is not None else incumbent_path.parent
    if incumbent_path.resolve().parent != root.resolve() or expert_path.resolve().parent != root.resolve():
        raise ValueError("Both prediction files must belong to the exact working directory")
    _, incumbent = read_submission(incumbent_path, labels, uids)
    backup = root / "p010_incumbent.csv"
    stage = root / "p010_candidate.partial.csv"
    if backup.exists() or stage.exists():
        raise FileExistsError("Preserve earlier candidate or baseline evidence")
    payload = incumbent_path.read_bytes()
    with backup.open("xb") as handle:
        handle.write(payload)
    receipt = {
        "schema_version": "p010_fixed_rank10_v1",
        "studies": len(uids),
        "incumbent_sha256": hashlib.sha256(payload).hexdigest(),
        "coefficients_predeclared": [0.9, 0.1],
        "labels": list(labels),
        "rank_ties": "average",
        "normalisation": "rank_1_to_N_divided_by_N",
        "accuracy_improvement_measured": False,
    }
    if len(uids) < 2:
        receipt.update(
            {"expert_used": False, "reason": "fewer_than_two_studies", "final_sha256": receipt["incumbent_sha256"]}
        )
        return receipt
    _, expert = read_submission(expert_path, labels, uids)
    result = fixed_rank10(incumbent, expert)
    write_submission_exclusive(stage, uids, result, labels)
    read_submission(stage, labels, uids)
    os.replace(stage, incumbent_path)
    receipt.update(
        {
            "expert_used": True,
            "expert_sha256": hashlib.sha256(expert_path.read_bytes()).hexdigest(),
            "final_sha256": hashlib.sha256(incumbent_path.read_bytes()).hexdigest(),
        }
    )
    return receipt
```

```python
%%writefile /kaggle/working/p010_src/p010_author224_metrics.py
"""Source-bound author224 branch; no downloads, training or remote operations.

The extracted numerical functions remain literal author source.  This wrapper
qualifies their use, isolates their globals, and rejects stock FP32 fallback.
Artificial CPU/CUDA execution can never qualify for production.
"""

from __future__ import annotations

import ast
import hashlib
import importlib
import json
from collections.abc import Mapping
from pathlib import Path
from typing import Any

AUTHOR_NOTEBOOK_SHA256 = "b80d3b72a1c34f2b39b3ea0d2952f039b8d27c3feb5c4db0a7d0ab62ae75354e"
AUTHOR_NORMALIZED_SHA256 = "993c756c94a8b7d04e91cc0cc568c845f60593c84d6a23b0a5e70419100f1d06"
CHECKPOINT_SHA256 = "3394fdd7e668286956e85af2554bef0386df5cd79d1c1af223b095c828b815d6"
CHECKPOINT_BYTES = 292842755
CHECKPOINT_FILENAME = "raptor_ft_alldata_t16_blendjev_oai_d96_c08_r224_swa.pt"
ARCH = "coatnet_rmlp_2_rw_224.sw_in12k_ft_in1k"
RESOLUTION = 224
LABELS = (
    "ACL", "MCL", "Medial Meniscus", "Lateral Meniscus", "Medial OA", "Lateral OA",
    "PF OA", "Effusion", "Synovitis", "Baker's", "Contusion", "Fracture",
)
SYMBOL_SHA256 = {
    "_crop": "1abff48888d855b653320b23f01ed0618bad29adc565a63e2061c1d34a0615ca",
    "build_backbone": "f5a1634d2e2a54a81265cf1e5b809afaf28b145d88d2cd24bcc48f5bb08ef1a3",
    "RaptorClassifier": "da765dd0bd9b371e563a541d5ffbc445d1138d2320d99cc0d8c2b7275e588b75",
    "load_model": "17a8beda51380fe0f0983b8bc7061558187ae112a10f0af7523bd49aecc1a3d5",
    "_eval_centers": "4ddfae29b35a0580a8b1df43bb4dcce0cac15b6a4d6473313d23390780d1acb3",
    "eval_windows": "6b5b2b026d9dd0e3396fc388e77237dcab76c7bc1ca49561337b30bfb704c9c4",
    "gpu_windows": "aa59baadaf18b1b276d7b6c99cd1680291e3e8fe0d6c12cbe8e5e2efe501d0a7",
    "infer_probs": "1185858b02024a7653421dda7b560bdc67eff06bf7fbe006582c9605cc1eb470",
    "rankpct": "e4ec008a6db3f7ada25c9968d72adbd47d1902a9933585610e20fc4326b12e34",
}
_CONSTANT_NAMES = {"IMG", "CROP_MM", "SLOTS", "MAXS", "K_EVAL", "NORM", "LAB", "CROP_FRAC",
                   "_MEAN", "_STD", "ARMS", "_WLUT"}
_IMPORT_TEXT = {
    "import os, sys, glob, time, json, gc", "import numpy as np",
    "import torch, torch.nn as nn, torch.nn.functional as F",
}


class Expert224ContractError(ValueError):
    """A source, checkpoint or invocation differs from the qualified contract."""


def sha256_file(path: str | Path) -> str:
    with Path(path).open("rb") as handle:
        return hashlib.file_digest(handle, "sha256").hexdigest()


def _text(cell: Mapping[str, Any]) -> str:
    source = cell.get("source")
    if isinstance(source, str):
        return source
    if isinstance(source, list) and all(isinstance(line, str) for line in source):
        return "".join(source)
    raise Expert224ContractError("Notebook source must be text or a list of text lines")


def _read_author(author_notebook: str | Path | Mapping[str, Any]) -> dict[str, Any]:
    if isinstance(author_notebook, Mapping):
        book = dict(author_notebook)
    else:
        path = Path(author_notebook)
        if sha256_file(path) != AUTHOR_NOTEBOOK_SHA256:
            raise Expert224ContractError("Author savedV2 Notebook SHA differs")
        book = json.loads(path.read_text(encoding="utf-8"))
    normalized = json.dumps(book, sort_keys=True, separators=(",", ":"), ensure_ascii=False).encode("utf-8")
    if hashlib.sha256(normalized).hexdigest() != AUTHOR_NORMALIZED_SHA256:
        raise Expert224ContractError("Author savedV2 Notebook content differs")
    return book


def _literal_with_decorators(source: str, node: ast.AST) -> str:
    start = node.lineno
    if isinstance(node, (ast.FunctionDef, ast.ClassDef, ast.AsyncFunctionDef)) and node.decorator_list:
        start = min(start, *(decorator.lineno for decorator in node.decorator_list))
    lines = source.splitlines(keepends=True)
    return "".join(lines[start - 1:node.end_lineno]).rstrip("\r\n")


def build_stock_source(author_notebook: str | Path | Mapping[str, Any]) -> str:
    """Return literal constants/imports and nine definitions from exact savedV2.

    The stock ``import timm`` is provided by ``make_stock_namespace`` so CPU
    tests can inject a tiny encoder without altering global import state.
    """
    book = _read_author(author_notebook)
    source = _text(book["cells"][4])
    nodes = ast.parse(source).body
    pieces = ["# Literal nartaa savedV2 author224 numerical source; timm is namespace-injected."]
    found_imports: set[str] = set()
    found_constants: set[str] = set()
    found_symbols: set[str] = set()
    for node in nodes:
        segment = ast.get_source_segment(source, node)
        if isinstance(node, (ast.Import, ast.ImportFrom)) and segment in _IMPORT_TEXT:
            found_imports.add(segment)
            pieces.append(_literal_with_decorators(source, node))
        elif isinstance(node, ast.Assign):
            names = {target.id for target in node.targets if isinstance(target, ast.Name)}
            if names & _CONSTANT_NAMES:
                if len(names) != 1 or not names <= _CONSTANT_NAMES or names & found_constants:
                    raise Expert224ContractError("Stock constant definition is ambiguous")
                found_constants.update(names)
                pieces.append(_literal_with_decorators(source, node))
        elif isinstance(node, (ast.FunctionDef, ast.ClassDef)) and node.name in SYMBOL_SHA256:
            if node.name in found_symbols or hashlib.sha256(segment.encode()).hexdigest() != SYMBOL_SHA256[node.name]:
                raise Expert224ContractError(f"Stock literal definition differs: {node.name}")
            if node.name == "infer_probs":
                if [ast.unparse(d) for d in node.decorator_list] != ["torch.no_grad()"]:
                    raise Expert224ContractError("Stock infer_probs no_grad decorator differs")
            found_symbols.add(node.name)
            pieces.append(_literal_with_decorators(source, node))
    if found_imports != _IMPORT_TEXT or found_constants != _CONSTANT_NAMES or found_symbols != set(SYMBOL_SHA256):
        raise Expert224ContractError("Required stock numerical imports/constants/symbols are incomplete")
    return "\n\n".join(pieces) + "\n"


def _source_contract(stock_source: str) -> str:
    """Validate an extracted source even when the original Notebook is absent."""
    tree = ast.parse(stock_source)
    found: dict[str, str] = {}
    constants: dict[str, ast.AST] = {}
    imports: set[str] = set()
    for node in tree.body:
        segment = ast.get_source_segment(stock_source, node)
        if isinstance(node, (ast.Import, ast.ImportFrom)):
            imports.add(segment)
        elif isinstance(node, ast.Assign):
            names = [target.id for target in node.targets if isinstance(target, ast.Name)]
            if len(names) != 1 or names[0] in constants:
                raise Expert224ContractError("Unexpected stock assignments")
            constants[names[0]] = node.value
        elif isinstance(node, (ast.FunctionDef, ast.ClassDef)):
            if node.name in found:
                raise Expert224ContractError("Duplicate stock numerical symbol")
            found[node.name] = hashlib.sha256(segment.encode()).hexdigest()
            expected_decorators = ["torch.no_grad()"] if node.name == "infer_probs" else []
            if [ast.unparse(d) for d in node.decorator_list] != expected_decorators:
                raise Expert224ContractError(f"Stock decorators differ: {node.name}")
        else:
            raise Expert224ContractError("Unexpected executable stock source node")
    if found != SYMBOL_SHA256 or set(constants) != _CONSTANT_NAMES or imports != _IMPORT_TEXT:
        raise Expert224ContractError("Extracted stock source does not match its bound definitions")
    # Constants are included in the full extracted-source digest, not accepted
    # merely because numerical functions happen to match.
    if hashlib.sha256(stock_source.encode()).hexdigest() != STOCK_SOURCE_SHA256:
        raise Expert224ContractError("Extracted stock constants/imports/source SHA differs")
    return STOCK_SOURCE_SHA256


# Filled from the exact savedV2 extraction; changing this requires a new audit.
STOCK_SOURCE_SHA256 = "b61046e344425c446370d3df1ca019cf278f13fba385117035f849484a5a85f0"


def make_stock_namespace(stock_source: str, *, timm_module: Any = None,
                         artificial_cpu: bool = False, artificial_device: str | None = None) -> dict[str, Any]:
    digest = _source_contract(stock_source)
    if artificial_cpu:
        if artificial_device not in (None, "cpu"):
            raise Expert224ContractError("Conflicting artificial device declarations")
        artificial_device = "cpu"
    if artificial_device not in (None, "cpu", "cuda"):
        raise Expert224ContractError("Artificial device must be explicitly cpu or cuda")
    artificial = artificial_device is not None
    if timm_module is not None and not artificial:
        raise Expert224ContractError("Injected encoder is restricted to artificial verification")
    if artificial and timm_module is None:
        raise Expert224ContractError("Artificial verification requires an injected tiny encoder")
    if timm_module is None:
        timm_module = importlib.import_module("timm")
    namespace: dict[str, Any] = {"__name__": "_p010_author224_stock", "timm": timm_module}
    exec(compile(stock_source, "<p010-author224-stock>", "exec"), namespace)
    if namespace["LAB"] != list(LABELS) or namespace["K_EVAL"] != 94 or namespace["IMG"] != 384:
        raise Expert224ContractError("Author target order or window dimensions differ")
    if namespace["NORM"] != "imagenet" or namespace["CROP_FRAC"] != 0.8 or namespace["MAXS"] != 96:
        raise Expert224ContractError("Author224 preprocessing constants differ")
    namespace["_p010_binding"] = {
        "stock_source_sha256": digest, "author_notebook_sha256": AUTHOR_NOTEBOOK_SHA256,
        "artificial_cpu": artificial_device == "cpu", "artificial_device": artificial_device,
        "artificial_mode": artificial, "production_qualified": not artificial,
        "namespace_identity": id(namespace),
    }
    namespace["_p010_bound_symbols"] = {name: namespace[name] for name in SYMBOL_SHA256}
    namespace["_p010_bound_modules"] = {name: namespace[name] for name in ("torch", "nn", "F", "np", "timm")}
    namespace["_p010_bound_lut"] = namespace["_WLUT"]
    namespace["_p010_bound_normalization"] = {name: namespace[name].clone() for name in ("_MEAN", "_STD")}
    return namespace


def _namespace_binding(namespace: dict[str, Any]) -> dict[str, Any]:
    binding = namespace.get("_p010_binding", {})
    if binding.get("stock_source_sha256") != STOCK_SOURCE_SHA256 or binding.get("namespace_identity") != id(namespace):
        raise Expert224ContractError("Stock namespace has no valid source binding")
    if any(namespace.get(name) is not value for name, value in namespace.get("_p010_bound_symbols", {}).items()):
        raise Expert224ContractError("A stock numerical callable was replaced after extraction")
    if set(namespace.get("_p010_bound_symbols", {})) != set(SYMBOL_SHA256):
        raise Expert224ContractError("Stock callable identity binding is incomplete")
    if any(namespace.get(name) is not value for name, value in namespace.get("_p010_bound_modules", {}).items()):
        raise Expert224ContractError("A stock dependency was replaced after extraction")
    if namespace.get("_WLUT") is not namespace.get("_p010_bound_lut"):
        raise Expert224ContractError("Stock expert LUT dictionary was replaced/shared with another branch")
    if set(namespace.get("_p010_bound_normalization", {})) != {"_MEAN", "_STD"} or any(
        not namespace["torch"].equal(namespace[name], expected)
        for name, expected in namespace["_p010_bound_normalization"].items()
    ):
        raise Expert224ContractError("Stock normalization tensors were changed after extraction")
    if (namespace["LAB"] != list(LABELS) or namespace["K_EVAL"] != 94 or namespace["IMG"] != 384
            or namespace["NORM"] != "imagenet" or namespace["CROP_FRAC"] != 0.8 or namespace["MAXS"] != 96
            or namespace["CROP_MM"] != 140.0
            or namespace["SLOTS"] != [("Sagittal", 1, 26), ("Sagittal", 0, 22),
                                      ("Coronal", 1, 18), ("Coronal", 0, 12), ("Axial", -1, 18)]):
        raise Expert224ContractError("Stock numerical constants were changed after extraction")
    return binding


def _device(namespace: dict[str, Any], device: Any):
    binding = _namespace_binding(namespace)
    torch = namespace["torch"]
    dev = torch.device(device)
    if binding["artificial_mode"]:
        if dev.type != binding["artificial_device"]:
            raise Expert224ContractError("Execution device differs from the explicit artificial device")
        if dev.type == "cuda" and not torch.cuda.is_available():
            raise Expert224ContractError("Artificial CUDA verification requires available CUDA")
    elif dev.type != "cuda" or not torch.cuda.is_available():
        raise Expert224ContractError("Production author224 requires CUDA before checkpoint loading")
    return dev


def load_expert224(path: str | Path, device: Any, *, stock_namespace: dict[str, Any],
                   artificial_expected_sha256: str | None = None):
    """Verify SHA/architecture/state, then invoke the unchanged strict loader."""
    namespace = stock_namespace
    binding = _namespace_binding(namespace)
    dev = _device(namespace, device)
    torch = namespace["torch"]
    path = Path(path)
    if artificial_expected_sha256 is not None and not binding["artificial_mode"]:
        raise Expert224ContractError("Production checkpoint SHA cannot be overridden")
    if binding["artificial_mode"] and artificial_expected_sha256 is None:
        raise Expert224ContractError("Artificial checkpoint requires its explicit test SHA")
    expected_sha = artificial_expected_sha256 if binding["artificial_mode"] else CHECKPOINT_SHA256
    actual_sha = sha256_file(path)
    if actual_sha != expected_sha:
        raise Expert224ContractError("Author224 checkpoint SHA differs")
    if not binding["artificial_mode"] and path.stat().st_size != CHECKPOINT_BYTES:
        raise Expert224ContractError("Author224 checkpoint file size differs")
    ck = torch.load(path, map_location="cpu", weights_only=False)
    if not isinstance(ck, Mapping) or ck.get("arch") != ARCH or ck.get("res") != RESOLUTION:
        raise Expert224ContractError("Explicit checkpoint arch/res do not match author224")
    state = ck.get("model")
    if not isinstance(state, Mapping) or not state or not all(isinstance(key, str) for key in state):
        raise Expert224ContractError("Author224 model state is missing or malformed")
    for key, value in state.items():
        if not isinstance(value, torch.Tensor):
            raise Expert224ContractError(f"Checkpoint state is not a tensor: {key}")
        if value.is_complex():
            raise Expert224ContractError(f"Checkpoint complex state is unsupported: {key}")
        if value.is_floating_point():
            if value.dtype != torch.float32 or not bool(torch.isfinite(value).all()):
                raise Expert224ContractError(f"Checkpoint floating state must be finite FP32: {key}")
    cls = state.get("clsW")
    if cls is None or cls.ndim != 2 or cls.shape[0] != 12 or cls.shape[1] < 1:
        raise Expert224ContractError("Checkpoint classifier must contain12 target rows")
    feature_dim = int(cls.shape[1])
    head_shapes = {"clsb": (12,), "norm.weight": (feature_dim,), "norm.bias": (feature_dim,),
                   "att.0.weight": (256, feature_dim), "att.0.bias": (256,),
                   "att.3.weight": (12, 256), "att.3.bias": (12,)}
    if any(key not in state or tuple(state[key].shape) != shape for key, shape in head_shapes.items()):
        raise Expert224ContractError("Checkpoint attention/head dimensions differ")
    label_fields = {key: ck[key] for key in ("labels", "label_order", "target_columns", "LAB") if key in ck}
    if any(not isinstance(value, (list, tuple)) or tuple(value) != LABELS for value in label_fields.values()):
        raise Expert224ContractError("Checkpoint explicit target metadata differs")
    state_count = len(state)
    del state, cls, ck
    model, resolution = namespace["load_model"](str(path), ARCH, RESOLUTION, str(dev), ngpu=1)
    if resolution != RESOLUTION or model.n != 12 or tuple(model.clsW.shape) != (12, feature_dim):
        raise Expert224ContractError("Stock strict loader returned unexpected architecture/head")
    if any(module.training for module in model.modules()):
        raise Expert224ContractError("Stock model is not completely in eval mode")
    if any(parameter.dtype != torch.float32 for parameter in model.parameters()):
        raise Expert224ContractError("Stock model parameters must remain FP32 for autocast")
    if any(parameter.device != dev for parameter in model.parameters()):
        # torch.device('cuda') has no index: compare against its resolved device.
        if dev.type != "cuda" or dev.index is not None or any(
            parameter.device.type != "cuda" or parameter.device.index != torch.cuda.current_device()
            for parameter in model.parameters()
        ):
            raise Expert224ContractError("Stock model is on an unexpected device")
    receipt = {
        "status": f"artificial_{dev.type}_passed" if binding["artificial_mode"] else "passed",
        "production_qualified": not binding["artificial_mode"], "checkpoint_sha256": actual_sha,
        "checkpoint_bytes": path.stat().st_size, "checkpoint_expected_sha256": expected_sha,
        "author_notebook_sha256": AUTHOR_NOTEBOOK_SHA256, "stock_source_sha256": STOCK_SOURCE_SHA256,
        "arch": ARCH, "resolution": resolution, "labels": list(LABELS), "head_outputs": 12,
        "state_tensors": state_count, "state_floating_dtype": "float32", "state_finite": True,
        "strict_load": True, "model_eval": True, "device": str(dev),
        "checkpoint_label_metadata_present": bool(label_fields),
        "checkpoint_label_metadata_fields": sorted(label_fields),
        "label_order_evidence": "literal author LAB and12-head; explicit checkpoint labels checked when present",
        "author_gold_selection_reuse": True,
        "author_gold_training_exclusion_independently_verified": False,
        "own_gold_fit_or_selection": False,
        "own_OOF_checkpoint": False,
        "author_provenance": "seed42 SWA15/13/12; publication card reports Gold58 selection/development reuse",
        "namespace_identity": id(namespace),
    }
    model._p010_expert224_binding = dict(receipt)
    return model, receipt


def _array_sha256(array) -> str:
    return hashlib.sha256(array.tobytes(order="C")).hexdigest()


def infer_expert224(stock_namespace: dict[str, Any], model: Any, vol: Any, mask: Any, device: Any):
    """Invoke exact windows/single stock forward with independent invocation gates."""
    namespace = stock_namespace
    binding = _namespace_binding(namespace)
    dev = _device(namespace, device)
    np, torch = namespace["np"], namespace["torch"]
    model_binding = getattr(model, "_p010_expert224_binding", {})
    if model_binding.get("namespace_identity") != id(namespace) or not model_binding.get("strict_load"):
        raise Expert224ContractError("Expert model is not bound to this stock strict loader")
    if not binding["artificial_mode"] and model_binding.get("checkpoint_sha256") != CHECKPOINT_SHA256:
        raise Expert224ContractError("Production expert model checkpoint binding differs")
    if not isinstance(vol, np.ndarray) or vol.shape != (96, 384, 384) or vol.dtype != np.uint8:
        raise Expert224ContractError("Shared volume must be96x384x384 uint8")
    if not isinstance(mask, np.ndarray) or mask.shape != (96,) or mask.dtype != np.uint8:
        raise Expert224ContractError("Shared mask must be96 uint8")
    if not bool(np.isin(mask, [0, 1]).all()):
        raise Expert224ContractError("Shared valid mask must contain only0/1")
    if any(module.training for module in model.modules()):
        raise Expert224ContractError("Expert eval mode was changed after loading")
    before = {"volume_sha256": _array_sha256(vol), "mask_sha256": _array_sha256(mask)}
    events: list[dict[str, Any]] = []
    counts = {"model": 0, "backbone": 0}

    def observe(kind: str, shape: tuple[int, ...]):
        def hook(module, args):
            counts[kind] += 1
            if counts[kind] != 1:
                # A ValueError bypasses the stock RuntimeError-to-FP32 handler.
                raise Expert224ContractError("Stock author224 retried/fell back instead of one forward")
            if len(args) != 1 or tuple(args[0].shape) != shape or args[0].dtype != torch.float32:
                raise Expert224ContractError(f"Stock {kind} batch shape/dtype differs")
            if module.training or torch.is_grad_enabled():
                raise Expert224ContractError("Expert forward must use eval and no_grad")
            autocast = bool(torch.is_autocast_enabled("cuda"))
            amp_dtype = str(torch.get_autocast_dtype("cuda")) if autocast else None
            if dev.type == "cuda" and (not autocast or amp_dtype != "torch.float16"):
                raise Expert224ContractError("CUDA stock forward did not use FP16 autocast")
            events.append({"kind": kind, "shape": list(shape), "input_dtype": str(args[0].dtype),
                           "training": bool(module.training), "grad_enabled": bool(torch.is_grad_enabled()),
                           "cuda_autocast": autocast, "autocast_dtype": amp_dtype})
        return hook

    handles = []
    try:
        handles.append(model.register_forward_pre_hook(observe("model", (1, 94, 3, 224, 224))))
        handles.append(model.backbone.register_forward_pre_hook(observe("backbone", (94, 3, 224, 224))))
        xwins = namespace["gpu_windows"](vol, mask, 94, RESOLUTION, str(dev))
        if tuple(xwins.shape) != (94, 3, 224, 224) or xwins.dtype != torch.float32:
            raise Expert224ContractError("Stock224 windows shape/dtype differs")
        if not bool(torch.isfinite(xwins).all()):
            raise Expert224ContractError("Stock224 windows contain nonfinite values")
        raw = namespace["infer_probs"](model, xwins, str(dev))
        if not isinstance(raw, np.ndarray) or raw.shape != (12,) or raw.dtype != np.float32:
            raise Expert224ContractError("Stock expert must return12 float32 raw probabilities")
        if not bool(np.isfinite(raw).all()) or not bool(((raw >= 0) & (raw <= 1)).all()):
            raise Expert224ContractError("Stock expert probabilities are nonfinite/out of range")
        if counts != {"model": 1, "backbone": 1}:
            raise Expert224ContractError("Stock expert did not execute exactly one full94 forward")
    finally:
        for handle in handles:
            handle.remove()
        # On success or error, shared input mutation is a fatal contract error.
        after = {"volume_sha256": _array_sha256(vol), "mask_sha256": _array_sha256(mask)}
        if before != after:
            raise Expert224ContractError("Expert mutated the shared volume/mask")
    receipt = {
        "status": f"artificial_{dev.type}_passed" if binding["artificial_mode"] else "passed",
        "production_qualified": not binding["artificial_mode"], "checkpoint_sha256": model_binding["checkpoint_sha256"],
        "author_notebook_sha256": AUTHOR_NOTEBOOK_SHA256, "stock_source_sha256": STOCK_SOURCE_SHA256,
        "arch": ARCH, "resolution": RESOLUTION, "labels": list(LABELS), "raw_outputs": 12,
        "raw_dtype": "float32", "finite": True, "in_range_0_1": True, "model_eval": True,
        "no_grad": True, "stock_forward_calls": 1, "backbone_forward_calls": 1,
        "backbone_batch_windows": 94, "views": 1, "fallback_calls": 0,
        "cuda_fp16_observed": dev.type == "cuda", "device": str(dev), "forward_events": events,
        "input_shape": [96, 384, 384], "input_dtype": "uint8", "mask_shape": [96],
        "mask_dtype": "uint8", "mask_valid_count": int(mask.sum()), "allmissing": not bool(mask.any()),
        "volume_mask_unchanged": True, "input_before": before, "input_after": after,
        "raw_probability_sha256": _array_sha256(raw),
        "author_gold_selection_reuse": True, "own_gold_fit_or_selection": False,
        "own_OOF_checkpoint": False,
    }
    return raw, receipt
```

```python
%%writefile /kaggle/working/p010_src/p010_author224_stock.py
# Literal nartaa savedV2 author224 numerical source; timm is namespace-injected.

import os, sys, glob, time, json, gc

import numpy as np

import torch, torch.nn as nn, torch.nn.functional as F

IMG = 384   # native 384 corpus: the model saw 384 px crops directly, no 336->384 upsample

CROP_MM = 140.0

SLOTS = [("Sagittal", 1, 26), ("Sagittal", 0, 22), ("Coronal", 1, 18),
         ("Coronal", 0, 12), ("Axial", -1, 18)]

MAXS = sum(s[2] for s in SLOTS)                     # 96

K_EVAL = 94   # every window position of the 96-slice test volume (full coverage)

NORM = "imagenet"

LAB = ["ACL", "MCL", "Medial Meniscus", "Lateral Meniscus", "Medial OA", "Lateral OA",
       "PF OA", "Effusion", "Synovitis", "Baker's", "Contusion", "Fracture"]

CROP_FRAC = 0.8   # centre crop of the field before the resize (train_knee RAPTOR_CROP_FRAC)

def _crop(t):
    H, W = t.shape[-2], t.shape[-1]; ch, cw = int(round(H * CROP_FRAC)), int(round(W * CROP_FRAC))
    y0, x0 = (H - ch) // 2, (W - cw) // 2
    return t[..., y0:y0 + ch, x0:x0 + cw]

_MEAN = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)

_STD = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)

ARMS = [   # round 2: native-res retrain with a centre crop of the 140 mm field
    {"file": "raptor_ft_alldata_t16_blendjev_oai_d96_c08_r224_swa.pt", "arch": "coatnet_rmlp_2_rw_224.sw_in12k_ft_in1k", "res": 224, "w": 1.0},
]

def build_backbone(arch, pretrained=False):
    # maxvit/maxxvit/coatnet are conv-attention hybrids: NO CLS token, NO interpolatable
    # pos-embed -> avg pool. The "vit" substring in "coatnet"/"maxvit" must NOT route them
    # down the ViT path (mirrors finetune_raptor.py exactly).
    hybrid = arch.startswith(("maxvit", "maxxvit", "coatnet", "coat_", "convnext"))
    is_vit = (not hybrid) and any(k in arch for k in ("vit", "deit", "dinov2", "eva", "beit"))
    kw = dict(pretrained=pretrained, num_classes=0, in_chans=3)
    if is_vit:
        kw.update(global_pool="token", dynamic_img_size=True)
    else:
        kw.update(global_pool="avg")
    return timm.create_model(arch, **kw)

class RaptorClassifier(nn.Module):
    def __init__(self, backbone, F_dim=768, n=12, drop=0.2):
        super().__init__()
        self.backbone = backbone
        self.norm = nn.LayerNorm(F_dim)
        self.att = nn.Sequential(nn.Linear(F_dim, 256), nn.Tanh(), nn.Dropout(drop),
                                 nn.Linear(256, n))
        self.clsW = nn.Parameter(torch.zeros(n, F_dim))
        self.clsb = nn.Parameter(torch.zeros(n))
        nn.init.trunc_normal_(self.clsW, std=0.02)
        self.n = n

    def encode(self, x):
        B, K = x.shape[:2]
        f = self.backbone(x.flatten(0, 1))
        return f.view(B, K, -1)

    def head(self, feats):
        h = self.norm(feats)
        a = self.att(h)
        a = torch.softmax(a, dim=1)
        pooled = torch.einsum("bkn,bkf->bnf", a, h)
        logits = (pooled * self.clsW).sum(-1) + self.clsb
        return logits

    def forward(self, x):
        return self.head(self.encode(x))

def load_model(pt_path, arch_default, res_default, device, ngpu=1):
    ck = torch.load(pt_path, map_location="cpu", weights_only=False)
    arch = ck.get("arch", arch_default)
    ck_res = int(ck.get("res", res_default))
    bb = build_backbone(arch, pretrained=False)
    model = RaptorClassifier(bb, F_dim=bb.num_features)
    model.load_state_dict(ck["model"], strict=True)
    model.eval().to(device)
    # NOTE: DataParallel removed on purpose. On the full hidden test it drove a system-RAM OOM
    # (per-forward module replication over many studies); a single T4 handles K_EVAL=24 windows
    # fine. Arms are also run SEQUENTIALLY (see main) so peak RAM == one model, not two.
    del ck
    gc.collect()
    return model, ck_res

def _eval_centers(mask, D, k):
    valid = np.where(mask > 0)[0]
    if len(valid) < 3:
        valid = np.arange(min(3, D))
    lo, hi = int(valid.min()), int(valid.max())
    cs = [c for c in range(lo + 1, hi) if c - 1 >= lo and c + 1 <= hi]
    if not cs:
        cs = [max(1, min((lo + hi) // 2, D - 2))]
    idx = np.linspace(0, len(cs) - 1, k).round().astype(int)
    return [cs[i] for i in idx]

def eval_windows(vol, mask, k, res, norm=NORM):
    D = vol.shape[0]
    cs = _eval_centers(mask, D, k)
    wins = np.empty((len(cs), 3, res, res), np.float32)
    for j, c in enumerate(cs):
        c = max(1, min(c, D - 2))
        tri = np.stack([vol[c - 1], vol[c], vol[c + 1]], 0).astype(np.float32) / 255.0
        t = _crop(torch.from_numpy(tri)) if CROP_FRAC < 1.0 else torch.from_numpy(tri)
        if t.shape[-1] != res:
            t = F.interpolate(t[None], size=(res, res), mode="bilinear",
                              align_corners=False)[0]
        wins[j] = t.numpy()
    x = torch.from_numpy(wins)
    if norm == "imagenet":
        x = (x - _MEAN) / _STD
    return x

_WLUT = {}

def gpu_windows(vol, mask, k, res, dev, norm=NORM):
    """Same windows as eval_windows, built on the GPU from the uint8 volume. Values come from a per-channel
    256-entry table computed on the CPU with eval_windows' own float32 ops, so they are bit-identical."""
    D = vol.shape[0]
    cs = np.array([max(1, min(c, D - 2)) for c in _eval_centers(mask, D, k)])
    tri = torch.from_numpy(np.ascontiguousarray(vol[np.stack([cs - 1, cs, cs + 1], 1)])).to(dev)   # (k,3,IMG,IMG) uint8
    key = (str(dev), norm)
    if key not in _WLUT:
        v = torch.from_numpy((np.arange(256, dtype=np.uint8).astype(np.float32) / 255.0))
        t = torch.stack([v, v, v], 0).view(3, 256, 1)
        if norm == 'imagenet':
            t = (t - _MEAN.view(3, 1, 1)) / _STD.view(3, 1, 1)
        _WLUT[key] = t.view(3, 256).contiguous().to(dev)
    L = _WLUT[key]
    x = torch.stack([L[c][tri[:, c].long()] for c in range(3)], 1)                   # exact float32 values
    if x.shape[-1] != res or CROP_FRAC < 1.0:
        if norm == 'imagenet':   # crop + resize BEFORE normalisation, as in eval_windows
            L0 = _WLUT.setdefault((str(dev), 'none'), torch.from_numpy(np.arange(256, dtype=np.uint8).astype(np.float32) / 255.0).to(dev))
            x = L0[tri.long()]
            if CROP_FRAC < 1.0: x = _crop(x)
            x = F.interpolate(x, size=(res, res), mode='bilinear', align_corners=False)
            x = (x - _MEAN.to(dev)) / _STD.to(dev)
        else:
            x = F.interpolate(_crop(x) if CROP_FRAC < 1.0 else x, size=(res, res), mode='bilinear', align_corners=False)
    return x

@torch.no_grad()
def infer_probs(model, xwins, device):
    x = xwins.unsqueeze(0).to(device)
    use_cuda = device != "cpu" and str(device).startswith("cuda")
    if use_cuda:
        # fp16 conv on T4 is fully cuDNN-supported (bf16 is NOT -> "no engine").
        try:
            with torch.autocast("cuda", dtype=torch.float16):
                o = torch.sigmoid(model(x).float())
            return o[0].cpu().numpy()
        except RuntimeError:
            # fp32 always has a Turing conv engine; slower but never drops a study.
            torch.cuda.empty_cache()
            o = torch.sigmoid(model(x).float())
            return o[0].cpu().numpy()
    o = torch.sigmoid(model(x).float())
    return o[0].cpu().numpy()

def rankpct(x):                                   # per-column percentile rank in [0,1]
    order = x.argsort(0).argsort(0).astype(np.float64)
    return order / max(1, (x.shape[0] - 1))
```

```python
%%writefile /kaggle/working/own_src/preprocess.py
"""Turn each DICOM series into a canonical uint8 volume [n_slices, SIZE, SIZE].

Canonical means: fixed physical pixel size, centre crop to a fixed field of view,
standard in-plane orientation, standard slice order and a per-series intensity window.
The same `load_series` is used for the training cache and for inference.

Usage: python src/preprocess.py [train|test] [n_series_limit]
"""
import os
import sys
from multiprocessing import Pool

import cv2
import numpy as np
import pandas as pd
import pydicom

DATA = "/mnt/ssd/rsna_knee/data"
WORK = "/mnt/ssd/rsna_knee/work"

MM_PER_PX = 0.4
SIZE = 384  # 153.6 mm field of view
MAX_SLICES = 64

# plane -> (normal axis, slice-order sign, (col axis, sign), (row axis, sign)) in LPS coordinates
# Sagittal: slices right->left, image columns anterior->posterior, rows superior->inferior
# Coronal:  slices anterior->posterior, columns patient-right->left, rows superior->inferior
# Axial:    slices superior->inferior, columns patient-right->left, rows anterior->posterior
CANON = {
    0: (+1, (1, +1), (2, -1)),
    1: (+1, (0, +1), (2, -1)),
    2: (-1, (0, +1), (1, +1)),
}
PLANES = ["Sagittal", "Coronal", "Axial"]


def _read(path):
    ds = pydicom.dcmread(path)
    arr = ds.pixel_array.astype(np.float32)
    if ds.get("PhotometricInterpretation", "MONOCHROME2") == "MONOCHROME1":
        arr = arr.max() - arr
    slope, inter = ds.get("RescaleSlope"), ds.get("RescaleIntercept")
    if slope is not None and inter is not None:
        arr = arr * float(slope) + float(inter)
    iop = [float(v) for v in ds.ImageOrientationPatient] if "ImageOrientationPatient" in ds else None
    ipp = [float(v) for v in ds.ImagePositionPatient] if "ImagePositionPatient" in ds else None
    ps = [float(v) for v in ds.PixelSpacing] if "PixelSpacing" in ds else None
    inst = float(ds.get("InstanceNumber") or 0)
    frames = list(arr) if arr.ndim == 3 else [arr]
    return [(f, iop, ipp, ps, inst + i * 1e-3) for i, f in enumerate(frames)]


def _canon_slice(img, iop, ps, plane):
    """Reorient one slice to the canonical in-plane orientation; returns (img, (row_mm, col_mm))."""
    ps = ps or [MM_PER_PX * img.shape[0] / SIZE] * 2  # no spacing: assume the image spans the crop
    if iop is None:
        return img, ps
    r, c = np.array(iop[:3]), np.array(iop[3:])
    _, (cax, csgn), (rax, rsgn) = CANON[plane]
    if abs(r[cax]) < abs(r[rax]):  # rows and columns are swapped relative to canonical
        img, r, c, ps = img.T, c, r, ps[::-1]
    if r[cax] * csgn < 0:
        img = img[:, ::-1]
    if c[rax] * rsgn < 0:
        img = img[::-1]
    return img, ps


def _resample(img, ps):
    """Centre crop to SIZE*MM_PER_PX mm, resample to MM_PER_PX, zero-pad to SIZE x SIZE."""
    out = np.zeros((SIZE, SIZE), np.float32)
    h, w = img.shape
    ch = min(h, int(round(SIZE * MM_PER_PX / ps[0])))
    cw = min(w, int(round(SIZE * MM_PER_PX / ps[1])))
    y0, x0 = (h - ch) // 2, (w - cw) // 2
    crop = np.ascontiguousarray(img[y0:y0 + ch, x0:x0 + cw])
    nh = max(1, min(SIZE, int(round(ch * ps[0] / MM_PER_PX))))
    nw = max(1, min(SIZE, int(round(cw * ps[1] / MM_PER_PX))))
    interp = cv2.INTER_AREA if nh < ch else cv2.INTER_LINEAR
    res = cv2.resize(crop, (nw, nh), interpolation=interp)
    oy, ox = (SIZE - nh) // 2, (SIZE - nw) // 2
    out[oy:oy + nh, ox:ox + nw] = res
    return out


def load_series(sdir, plane_name=None):
    """Returns (uint8 volume [n, SIZE, SIZE], info dict). Unreadable slices are skipped."""
    slices, bad = [], 0
    for fn in os.listdir(sdir):
        try:
            slices.extend(_read(os.path.join(sdir, fn)))
        except Exception:
            bad += 1
    info = {"n_raw": len(slices), "n_bad": bad}
    if not slices:
        return np.zeros((1, SIZE, SIZE), np.uint8), info

    iop = next((s[1] for s in slices if s[1] is not None), None)
    if iop is not None:
        normal = np.cross(iop[:3], iop[3:])
        plane = int(np.abs(normal).argmax())
    else:
        normal, plane = None, PLANES.index(plane_name) if plane_name in PLANES else 0
    info["plane"] = PLANES[plane]

    if normal is not None and all(s[2] is not None for s in slices):
        sgn = CANON[plane][0] * np.sign(normal[plane])
        slices.sort(key=lambda s: float(np.dot(s[2], normal)) * sgn)
        pos = [float(np.dot(s[2], normal)) for s in slices]
        info["slice_gap"] = float(np.median(np.abs(np.diff(pos)))) if len(pos) > 1 else 0.0
    else:
        slices.sort(key=lambda s: s[4])
    if len(slices) > MAX_SLICES:
        keep = np.linspace(0, len(slices) - 1, MAX_SLICES).round().astype(int)
        slices = [slices[i] for i in keep]

    vol = np.stack([_resample(*_canon_slice(s[0], s[1], s[3], plane)) for s in slices])
    lo, hi = np.percentile(vol[:, ::4, ::4], [1, 99.5])
    vol = np.clip((vol - lo) / max(hi - lo, 1e-6), 0, 1)
    info["n"] = len(vol)
    return (vol * 255).round().astype(np.uint8), info


def _job(args):
    split, study, series, plane, out_dir = args
    out = f"{out_dir}/{series}.npy"
    info = {"study": study, "series": series}
    if os.path.exists(out):
        return None
    try:
        vol, extra = load_series(f"{DATA}/{split}_series/{study}/{series}", plane)
        np.save(out + ".tmp.npy", vol)
        os.replace(out + ".tmp.npy", out)
        info.update(extra)
    except Exception as e:
        info["error"] = f"{type(e).__name__}: {e}"[:200]
    return info


def main():
    split = sys.argv[1] if len(sys.argv) > 1 else "train"
    limit = int(sys.argv[2]) if len(sys.argv) > 2 else None
    out_dir = f"{WORK}/cache{SIZE}_{split}"
    os.makedirs(out_dir, exist_ok=True)
    se = pd.read_csv(f"{DATA}/{split}_series.csv")
    if limit:
        se = se.sample(limit, random_state=0)
    tasks = [(split, a, b, p, out_dir) for a, b, p in zip(se.StudyInstanceUID, se.SeriesInstanceUID, se.Anatomical_Plane)]
    rows = []
    with Pool(18) as p:
        for i, r in enumerate(p.imap_unordered(_job, tasks, chunksize=4)):
            if r:
                rows.append(r)
            if i % 1000 == 0:
                print(f"{i}/{len(tasks)}", flush=True)
    df = pd.DataFrame(rows)
    meta = f"{WORK}/cache{SIZE}_{split}_meta.csv"
    if os.path.exists(meta) and len(df):
        df = pd.concat([pd.read_csv(meta), df]).drop_duplicates("series", keep="last")
    if len(df):
        df.to_csv(meta, index=False)
        print("done", len(df), "errors:", int(df["error"].notna().sum()) if "error" in df else 0)


if __name__ == "__main__":
    main()
```

```python
%%writefile /kaggle/working/own_src/knee.py
"""Shared dataset and model code for training and inference."""
import math

import numpy as np
import timm
import torch
import torch.nn as nn
import torch.nn.functional as F

LABELS = ["ACL", "MCL", "Medial Meniscus", "Lateral Meniscus", "Medial OA", "Lateral OA",
          "PF OA", "Effusion", "Synovitis", "Baker's", "Contusion", "Fracture"]
PLANES = ["Sagittal", "Coronal", "Axial"]
# slot = plane index + 3 * (1 - fat_suppressed): fat-suppressed planes first
N_SLOTS = 6
MEAN = (0.485, 0.456, 0.406)
STD = (0.229, 0.224, 0.225)


def slot_of(plane, fat_sup):
    return PLANES.index(plane) + 3 * (1 - int(fat_sup))


def build_study_table(series_df, n_slices):
    """series_df: StudyInstanceUID, SeriesInstanceUID, Fat_Suppression, Anatomical_Plane.
    n_slices: dict series -> cached slice count. Returns {study: [[series ids of slot 0], ...]}
    with each slot's candidates sorted by slice count (largest first)."""
    table = {}
    for st, sr, fs, pl in zip(series_df.StudyInstanceUID, series_df.SeriesInstanceUID,
                              series_df.Fat_Suppression, series_df.Anatomical_Plane):
        if n_slices.get(sr, 0) < 3:
            continue
        table.setdefault(st, [[] for _ in range(N_SLOTS)])[slot_of(pl, fs)].append(sr)
    for slots in table.values():
        for cands in slots:
            cands.sort(key=lambda s: -n_slices[s])
        # Intelligent anatomical plane fallback: if fat-sat is missing, use non-fat-sat (and vice-versa)
        for p in range(3):
            fs_s, nonfs_s = p, p + 3
            if not slots[fs_s] and slots[nonfs_s]:
                slots[fs_s] = list(slots[nonfs_s])
            elif not slots[nonfs_s] and slots[fs_s]:
                slots[nonfs_s] = list(slots[fs_s])
    return table


def window_centres(n, k, train, lo=0.04, hi=0.96):
    """k slice indices (centres of 3-slice windows) spread over the series."""
    a, b = max(1.0, lo * (n - 1)), min(n - 2.0, hi * (n - 1))
    if b < a:
        a = b = (n - 1) / 2
    if train:
        edges = np.linspace(a, b, k + 1)
        c = edges[:-1] + np.random.rand(k) * (edges[1:] - edges[:-1])
    else:
        c = np.linspace(a, b, k)
    return np.clip(np.round(c).astype(int), 1, max(1, n - 2))


class StudyDataset(torch.utils.data.Dataset):
    """One item = one study: uint8 windows [N_SLOTS, k, 3, H, W], slot mask, slice positions, target."""

    def __init__(self, studies, table, cache_dir, targets=None, k=8, train=False, loader=None):
        self.studies, self.table, self.cache_dir = list(studies), table, cache_dir
        self.targets, self.k, self.train = targets, k, train
        self.loader = loader or (lambda s: np.load(f"{self.cache_dir}/{s}.npy", mmap_mode="r"))

    def __len__(self):
        return len(self.studies)

    def __getitem__(self, i):
        st = self.studies[i]
        slots = self.table.get(st, [[] for _ in range(N_SLOTS)])
        x, mask, pos = None, np.zeros(N_SLOTS, bool), np.zeros((N_SLOTS, self.k), np.float32)
        for s, cands in enumerate(slots):
            if not cands:
                continue
            sr = cands[np.random.randint(len(cands))] if self.train else cands[0]
            vol = self.loader(sr)
            n = len(vol)
            c = window_centres(n, self.k, self.train)
            win = np.stack([vol[np.clip([j - 1, j, j + 1], 0, n - 1)] for j in c])  # [k, 3, H, W]
            if x is None:
                x = np.zeros((N_SLOTS, self.k) + win.shape[1:], np.uint8)
            x[s], mask[s], pos[s] = win, True, c / max(n - 1, 1)
        if x is None:
            x = np.zeros((N_SLOTS, self.k, 3, 384, 384), np.uint8)
        y = self.targets[i] if self.targets is not None else np.zeros(len(LABELS), np.float32)
        return torch.from_numpy(x), torch.from_numpy(mask), torch.from_numpy(pos), torch.from_numpy(np.asarray(y, np.float32))


def make_views(x, res, train, group, fov=0.92):
    """x: uint8 [N, 3, H, W] with N = n_series * group. Crops a centred `fov` fraction of the cached
    field of view and resizes to res; in training adds rotation/scale/shift and gain/gamma jitter
    that is shared within a series."""
    n = x.shape[0]
    ns = n // group
    dev = x.device
    x = x.float().div_(255.0)
    if train:
        ang = (torch.rand(ns, device=dev) - 0.5) * 2 * math.radians(12)
        sc = fov * (1 + (torch.rand(ns, device=dev) - 0.5) * 0.24)
        tx = (torch.rand(ns, device=dev) - 0.5) * 0.12
        ty = (torch.rand(ns, device=dev) - 0.5) * 0.12
        gamma = torch.exp((torch.rand(ns, device=dev) - 0.5) * 0.5)
        gain = 1 + (torch.rand(ns, device=dev) - 0.5) * 0.2
    else:
        ang = torch.zeros(ns, device=dev)
        sc = torch.full((ns,), fov, device=dev)
        tx = ty = torch.zeros(ns, device=dev)
        gamma = gain = None
    cos, sin = torch.cos(ang) * sc, torch.sin(ang) * sc
    theta = torch.stack([torch.stack([cos, -sin, tx], 1), torch.stack([sin, cos, ty], 1)], 1)
    theta = theta.repeat_interleave(group, 0)
    grid = F.affine_grid(theta, (n, 3, res, res), align_corners=False)
    if x.shape[-1] > 1.5 * res:  # avoid aliasing when shrinking a lot
        x = F.avg_pool2d(x, 2) if x.shape[-1] >= 2 * res else F.interpolate(x, scale_factor=0.75, mode="bilinear", antialias=True)
    x = F.grid_sample(x, grid, mode="bilinear", padding_mode="zeros", align_corners=False)
    if train:
        g = gamma.repeat_interleave(group).view(n, 1, 1, 1)
        x = x.clamp_min(1e-4).pow(g) * gain.repeat_interleave(group).view(n, 1, 1, 1)
    mean = torch.tensor(MEAN, device=dev).view(1, 3, 1, 1)
    std = torch.tensor(STD, device=dev).view(1, 3, 1, 1)
    return (x - mean) / std


class KneeNet(nn.Module):
    """2D backbone on 3-slice windows -> tokens tagged with series slot and slice position ->
    small transformer over all windows of the study -> one attention pooling per finding."""

    def __init__(self, backbone="convnext_tiny", pretrained=True, dim=384, depth=2, drop=0.2, drop_path=0.1,
                 img_size=None):
        super().__init__()
        kw = {"drop_path_rate": drop_path} if drop_path else {}
        if img_size and backbone.startswith(("vit_", "eva", "beit")):  # fixed-grid transformers need the input size
            kw["img_size"] = img_size
        self.enc = timm.create_model(backbone, pretrained=pretrained, num_classes=0, **kw)
        self.proj = nn.Sequential(nn.Dropout(drop), nn.Linear(self.enc.num_features, dim), nn.LayerNorm(dim))
        self.slot_emb = nn.Embedding(N_SLOTS, dim)
        self.pos_emb = nn.Sequential(nn.Linear(1, dim), nn.GELU(), nn.Linear(dim, dim))
        layer = nn.TransformerEncoderLayer(dim, 8, dim * 2, dropout=0.1, batch_first=True, norm_first=True)
        self.tf = nn.TransformerEncoder(layer, depth, enable_nested_tensor=False) if depth else None
        self.att = nn.Sequential(nn.Linear(dim, 256), nn.Tanh(), nn.Dropout(drop), nn.Linear(256, len(LABELS)))
        self.cls_w = nn.Parameter(torch.randn(len(LABELS), dim) * 0.02)
        self.cls_b = nn.Parameter(torch.zeros(len(LABELS)))

    def head(self, feat, mask, pos):
        """feat [B, S, K, D], mask [B, S] bool, pos [B, S, K] in 0..1 -> logits [B, 12]."""
        b, s, k, _ = feat.shape
        slot = torch.arange(s, device=feat.device).view(1, s, 1).expand(b, s, k)
        h = self.proj(feat) + self.slot_emb(slot) + self.pos_emb(pos.unsqueeze(-1).to(feat.dtype))
        h = h.view(b, s * k, -1)
        pad = ~mask.unsqueeze(-1).expand(b, s, k).reshape(b, s * k)
        pad = pad & ~pad.all(1, keepdim=True)  # a study with no series at all attends to everything
        if self.tf is not None:
            h = self.tf(h, src_key_padding_mask=pad)
        a = self.att(h).float().masked_fill(pad.unsqueeze(-1), -1e4).softmax(1)  # [B, T, C]
        pooled = torch.einsum("btc,btd->bcd", a.to(h.dtype), h)
        return (pooled * self.cls_w).sum(-1) + self.cls_b

    def forward(self, x, mask, pos, res=256, train=False):
        """x uint8 [B, S, K, 3, H, W]."""
        b, s, k = x.shape[:3]
        keep = mask.view(-1)  # run the backbone only on series that exist
        xv = x.view(b * s, k, *x.shape[3:])[keep].flatten(0, 1)
        f = self.enc(make_views(xv, res, train, k).contiguous(memory_format=torch.channels_last))
        feat = f.new_zeros(b * s, k, f.shape[-1])
        feat[keep] = f.view(-1, k, f.shape[-1])
        return self.head(feat.view(b, s, k, -1), mask, pos)
```

```python
%%writefile /kaggle/working/own_src/infer.py
"""Predict the 12 findings for every study straight from DICOM folders.

Used by the Kaggle submission notebook and locally to check that predictions from raw DICOMs
match those from the training cache.

Usage: python src/infer.py <data_dir> <split> <out_csv> <ckpt> [<ckpt> ...]
  data_dir contains <split>.csv, <split>_series.csv and <split>_series/
"""
import os
import sys

import numpy as np
import pandas as pd
import torch

from knee import LABELS, N_SLOTS, KneeNet, build_study_table, window_centres
from preprocess import MAX_SLICES, SIZE, load_series


class DicomStudyDataset(torch.utils.data.Dataset):
    """Loads every selected series of a study once; windows are cut per model setting later."""

    def __init__(self, studies, table, root, k):
        self.studies, self.table, self.root, self.k = list(studies), table, root, k

    def __len__(self):
        return len(self.studies)

    def __getitem__(self, i):
        st = self.studies[i]
        slots = self.table.get(st, [[] for _ in range(N_SLOTS)])
        x = np.zeros((N_SLOTS, self.k, 3, SIZE, SIZE), np.uint8)
        mask, pos = np.zeros(N_SLOTS, bool), np.zeros((N_SLOTS, self.k), np.float32)
        for s, cands in enumerate(slots):
            if not cands:
                continue
            try:
                vol, _ = load_series(f"{self.root}/{st}/{cands[0]}")
            except Exception:
                continue
            n = len(vol)
            if n < 3:
                continue
            c = window_centres(n, self.k, False)
            x[s] = np.stack([vol[np.clip([j - 1, j, j + 1], 0, n - 1)] for j in c])
            mask[s], pos[s] = True, c / max(n - 1, 1)
        return torch.from_numpy(x), torch.from_numpy(mask), torch.from_numpy(pos)


def load_models(ckpts, device="cuda"):
    models = []
    for path in ckpts:
        ck = torch.load(path, map_location="cpu", weights_only=False)
        # the pretrained tag after the dot is irrelevant here and unknown to older timm versions
        m = KneeNet(ck["args"]["backbone"].split(".")[0], pretrained=False, img_size=ck["args"]["res"])
        m.load_state_dict(ck["model"])
        models.append((m.to(device).to(memory_format=torch.channels_last).eval(), ck["args"]["res"]))
    return models


@torch.no_grad()
def run(data_dir, split, ckpts, k=16, studies=None, workers=4, bs=4):
    se = pd.read_csv(f"{data_dir}/{split}_series.csv")
    root = f"{data_dir}/{split}_series"
    if studies is None:
        studies = pd.read_csv(f"{data_dir}/{split}.csv").StudyInstanceUID.tolist()
    se = se[se.StudyInstanceUID.isin(set(studies))]
    n_files = {}
    for st, sr in zip(se.StudyInstanceUID, se.SeriesInstanceUID):
        d = f"{root}/{st}/{sr}"
        n_files[sr] = min(len(os.listdir(d)), MAX_SLICES) if os.path.isdir(d) else 0
    table = build_study_table(se, n_files)
    dl = torch.utils.data.DataLoader(DicomStudyDataset(studies, table, root, k), batch_size=bs,
                                     num_workers=workers, pin_memory=True)
    models = load_models(ckpts)
    out = []
    for x, mask, pos in dl:
        x, mask, pos = x.cuda(non_blocking=True), mask.cuda(), pos.cuda()
        with torch.autocast("cuda", dtype=torch.float16):
            p = torch.stack([m(x, mask, pos, res=res).float().sigmoid() for m, res in models]).mean(0)
        out.append(p.cpu())
    return pd.DataFrame(torch.cat(out).numpy(), index=pd.Index(studies, name="StudyInstanceUID"), columns=LABELS)


if __name__ == "__main__":
    data_dir, split, out_csv, ckpts = sys.argv[1], sys.argv[2], sys.argv[3], sys.argv[4:]
    run(data_dir, split, ckpts).to_csv(out_csv)
    print("wrote", out_csv)
```
